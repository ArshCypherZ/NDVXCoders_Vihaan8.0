```python
import pandas as pd
import numpy as np
import os
import json
import joblib
import re
import warnings
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, OneHotEncoder, MultiLabelBinarizer
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.neighbors import LocalOutlierFactor
from sklearn.svm import OneClassSVM
from sklearn.pipeline import Pipeline
from sklearn.metrics import classification_report, confusion_matrix, f1_score, accuracy_score
from sklearn.preprocessing import FunctionTransformer


# Suppress warnings
warnings.filterwarnings('ignore')

# --- Configuration ---
BASE_DIR = os.getenv('BASE_DIR', 'archive')
OPTIONS_JSON_FILE = os.getenv('OPTIONS_JSON_FILE', 'frontend_options.json')
ARTIFACTS_FILE = os.getenv('ARTIFACTS_FILE', 'model_artifacts_tuned.joblib')

# Training Configuration
VALIDATION_SPLIT_SIZE = float(os.getenv('VALIDATION_SPLIT_SIZE', 0.2)) # Use same split for consistency
TFIDF_MAX_FEATURES = int(os.getenv('TFIDF_MAX_FEATURES', 1000))

# --- Hyperparameter Ranges to Test ---
LOF_CONTAMINATION_VALUES = [0.005, 0.01, 0.02, 0.05, 0.1, 0.15, 0.2] # Expanded LOF range
OCSVM_NU_VALUES = [0.05, 0.1, 0.15, 0.2, 0.25, 0.3, 0.4, 0.5] # Expanded OCSVM nu range
OCSVM_GAMMA_VALUES = ['scale', 'auto', 0.001, 0.01, 0.1, 0.5, 1.0] # Expanded OCSVM gamma range

# --- Helper Functions ---
# Helper function for cleaning text column within the pipeline
def clean_text_series(X):
    # X could be a Series, ndarray, or list passed by FunctionTransformer
    cleaned_X = []
    # Iterate through elements, handle NA, ensure string output
    for item in X:
        if pd.isna(item):
            cleaned_X.append('')
        else:
            cleaned_X.append(str(item))
    # Return a list, which TfidfVectorizer can handle
    return cleaned_X

def extract_dosage_numeric(desc):
    if pd.isna(desc): return np.nan
    match = re.search(r'(\d+(?:\.\d+)?)\s*(?:MG|MCG|ML)', str(desc), re.IGNORECASE)
    return float(match.group(1)) if match else np.nan

def extract_drug_name(desc):
     if pd.isna(desc): return "Unknown"
     pattern = re.compile(r'^([a-zA-Z\s-]+)(?:\s+\d|\s*\(|\s+MG|\s+MCG|\s+ML)', re.IGNORECASE)
     match = pattern.match(str(desc))
     if match:
         drug = match.group(1).strip().title()
         return drug if len(drug) > 3 else "Unknown"
     return "Unknown"

def safe_load_csv(path, usecols=None):
    try:
        return pd.read_csv(path, usecols=usecols, low_memory=False)
    except UnicodeDecodeError:
        try:
            return pd.read_csv(path, usecols=usecols, encoding='latin1', low_memory=False)
        except Exception as e: print(f"  Error loading {path} (latin1): {e}"); return None
    except FileNotFoundError: print(f"  File not found: {path}"); return None
    except ValueError as e: print(f"  Column error loading {path}: {e}"); return None
    except Exception as e: print(f"  Error loading {path}: {e}"); return None

# --- 1. Load and Preprocess Training Data  ---
print("--- Step 1: Loading and Preprocessing Training Data ---")
PATIENTS_COLS = ['patient', 'birthdate', 'gender']
ENCOUNTERS_COLS = ['ID', 'DATE', 'PATIENT', 'DESCRIPTION', 'REASONDESCRIPTION']
CONDITIONS_COLS = ['PATIENT', 'ENCOUNTER', 'DESCRIPTION', 'START', 'STOP']
MEDICATIONS_COLS = ['PATIENT', 'ENCOUNTER', 'DESCRIPTION', 'START', 'STOP']
PROCEDURES_COLS = ['PATIENT', 'ENCOUNTER', 'DESCRIPTION']
ALLERGIES_COLS = ['PATIENT', 'DESCRIPTION']
CLAIMS_COLS = ['PATIENT', 'ENCOUNTER', 'DIAGNOSIS']
OBSERVATIONS_COLS = ['PATIENT', 'ENCOUNTER', 'DATE', 'CODE', 'VALUE', 'UNITS', 'DESCRIPTION']

# Load data
df_patients = safe_load_csv(os.path.join(BASE_DIR, 'patients.csv'), PATIENTS_COLS)
df_encounters = safe_load_csv(os.path.join(BASE_DIR, 'encounters.csv'), ENCOUNTERS_COLS)
df_conditions = safe_load_csv(os.path.join(BASE_DIR, 'conditions.csv'), CONDITIONS_COLS)
df_medications = safe_load_csv(os.path.join(BASE_DIR, 'medications.csv'), MEDICATIONS_COLS)
df_procedures = safe_load_csv(os.path.join(BASE_DIR, 'procedures.csv'), PROCEDURES_COLS)
df_allergies = safe_load_csv(os.path.join(BASE_DIR, 'allergies.csv'), ALLERGIES_COLS)
df_claims = safe_load_csv(os.path.join(BASE_DIR, 'claims.csv'), CLAIMS_COLS)
df_observations = safe_load_csv(os.path.join(BASE_DIR, 'observations.csv'), OBSERVATIONS_COLS)

if df_patients is None or df_encounters is None:
    print("Error: Essential dataframes ('patients', 'encounters') could not be loaded. Exiting.")
    exit()

# Standardize IDs
df_patients = df_patients.rename(columns={'patient': 'PATIENT'})
df_encounters = df_encounters.rename(columns={'ID': 'ENCOUNTER'})
if df_conditions is not None: df_conditions = df_conditions.rename(columns={'patient': 'PATIENT', 'Id': 'ENCOUNTER', 'ID': 'ENCOUNTER'}, errors='ignore')
if df_medications is not None: df_medications = df_medications.rename(columns={'patient': 'PATIENT', 'Id': 'ENCOUNTER', 'ID': 'ENCOUNTER'}, errors='ignore')
if df_procedures is not None: df_procedures = df_procedures.rename(columns={'patient': 'PATIENT', 'Id': 'ENCOUNTER', 'ID': 'ENCOUNTER'}, errors='ignore')
if df_allergies is not None: df_allergies = df_allergies.rename(columns={'patient': 'PATIENT'}, errors='ignore')
if df_claims is not None: df_claims = df_claims.rename(columns={'patient': 'PATIENT', 'Id': 'ENCOUNTER', 'ID': 'CLAIM_ID'}, errors='ignore')
if df_observations is not None: df_observations = df_observations.rename(columns={'patient': 'PATIENT', 'Id': 'ENCOUNTER', 'ID': 'ENCOUNTER', 'DATE': 'OBS_DATE'}, errors='ignore')

# Aggregate allergies
patient_allergies = pd.DataFrame({'PATIENT': [], 'knownAllergies': []})
if df_allergies is not None and 'PATIENT' in df_allergies.columns and 'DESCRIPTION' in df_allergies.columns:
    patient_allergies = df_allergies.dropna(subset=['DESCRIPTION']).groupby('PATIENT')['DESCRIPTION'].apply(lambda x: list(set(str(s).strip().title() for s in x if str(s).strip()))).reset_index()
    patient_allergies = patient_allergies.rename(columns={'DESCRIPTION': 'knownAllergies'})

# Aggregate counts
encounter_proc_counts = pd.DataFrame({'PATIENT': [], 'ENCOUNTER': [], 'proc_count': []})
if df_procedures is not None and 'PATIENT' in df_procedures.columns and 'ENCOUNTER' in df_procedures.columns:
    encounter_proc_counts = df_procedures.groupby(['PATIENT', 'ENCOUNTER']).size().reset_index(name='proc_count')
encounter_claim_counts = pd.DataFrame({'PATIENT': [], 'ENCOUNTER': [], 'claim_count': []})
if df_claims is not None and 'PATIENT' in df_claims.columns and 'ENCOUNTER' in df_claims.columns:
    encounter_claim_counts = df_claims.groupby(['PATIENT', 'ENCOUNTER']).size().reset_index(name='claim_count')

# Merge data (simplified merge logic from train script)
merged_df = pd.merge(df_encounters, df_patients, on='PATIENT', how='left')
if df_medications is not None:
    med_cols_to_merge = ['PATIENT', 'ENCOUNTER', 'DESCRIPTION', 'START', 'STOP']
    df_medications_renamed = df_medications[med_cols_to_merge].rename(columns={'DESCRIPTION': 'DESCRIPTION_med', 'START': 'START_med', 'STOP': 'STOP_med'})
    merged_df = pd.merge(merged_df, df_medications_renamed, on=['PATIENT', 'ENCOUNTER'], how='left')
else:
    merged_df['DESCRIPTION_med'], merged_df['START_med'], merged_df['STOP_med'] = pd.NA, pd.NaT, pd.NaT
merged_df = pd.merge(merged_df, patient_allergies, on='PATIENT', how='left')
merged_df = pd.merge(merged_df, encounter_proc_counts, on=['PATIENT', 'ENCOUNTER'], how='left')
merged_df = pd.merge(merged_df, encounter_claim_counts, on=['PATIENT', 'ENCOUNTER'], how='left')
if df_conditions is not None:
    df_conditions_to_merge = df_conditions.rename(columns={'PATIENT': 'PATIENT_cond_ref', 'ENCOUNTER': 'ENCOUNTER_cond_ref'}, errors='ignore')
    merged_df = pd.merge(merged_df, df_conditions_to_merge, left_on=['PATIENT', 'ENCOUNTER'], right_on=['PATIENT_cond_ref', 'ENCOUNTER_cond_ref'], how='left')
if df_observations is not None:
     df_observations_to_merge = df_observations.rename(columns={'PATIENT': 'PATIENT_obs_ref', 'ENCOUNTER': 'ENCOUNTER_obs_ref'}, errors='ignore')
     merged_df = pd.merge(merged_df, df_observations_to_merge, left_on=['PATIENT', 'ENCOUNTER'], right_on=['PATIENT_obs_ref', 'ENCOUNTER_obs_ref'], how='left')

# Fill NaNs and initialize lists
merged_df['proc_count'] = merged_df['proc_count'].fillna(0).astype(int)
merged_df['claim_count'] = merged_df['claim_count'].fillna(0).astype(int)
merged_df['knownAllergies'] = merged_df['knownAllergies'].apply(lambda x: x if isinstance(x, list) else [])
merged_df['conditions_during_encounter'] = [[] for _ in range(len(merged_df))]
merged_df['new_conditions_after_med_start'] = [[] for _ in range(len(merged_df))]
merged_df['trialSideEffects'] = [[] for _ in range(len(merged_df))]
merged_df['aggregated_observations'] = [{} for _ in range(len(merged_df))]

# Feature Engineering (Dates, Age, Gender, Text, Drug/Dosage, Duration)
merged_df['birthdate'] = pd.to_datetime(merged_df['birthdate'], errors='coerce')
merged_df['DATE'] = pd.to_datetime(merged_df['DATE'], errors='coerce')
merged_df['START_med'] = pd.to_datetime(merged_df['START_med'], errors='coerce')
merged_df['STOP_med'] = pd.to_datetime(merged_df['STOP_med'], errors='coerce')
if 'OBS_DATE_obs' in merged_df.columns: merged_df['OBS_DATE_obs'] = pd.to_datetime(merged_df['OBS_DATE_obs'], errors='coerce')
if 'START_cond' in merged_df.columns: merged_df['START_cond'] = pd.to_datetime(merged_df['START_cond'], errors='coerce')
if 'STOP_cond' in merged_df.columns: merged_df['STOP_cond'] = pd.to_datetime(merged_df['STOP_cond'], errors='coerce')
valid_dates_mask = merged_df['birthdate'].notna() & merged_df['DATE'].notna()
merged_df['age'] = np.nan
merged_df.loc[valid_dates_mask, 'age'] = (merged_df.loc[valid_dates_mask, 'DATE'] - merged_df.loc[valid_dates_mask, 'birthdate']).dt.days // 365.25
merged_df['age'] = merged_df['age'].clip(0, 120)
if 'gender' in merged_df.columns: merged_df['gender'] = merged_df['gender'].map({'M': 'Male', 'F': 'Female'}).fillna('Other')
else: merged_df['gender'] = 'Other'
merged_df['doctorNotes'] = merged_df['DESCRIPTION'].fillna('') + ' | ' + merged_df['REASONDESCRIPTION'].fillna('')
merged_df['doctorNotes'] = merged_df['doctorNotes'].replace(r'^\s*\|\s*$', '', regex=True).str.strip().fillna("No notes")
merged_df['noteLength'] = merged_df['doctorNotes'].apply(len)
merged_df['noteWordCount'] = merged_df['doctorNotes'].apply(lambda x: len(x.split()))
merged_df['drugName'] = merged_df['DESCRIPTION_med'].apply(extract_drug_name)
merged_df['dosage_numeric'] = merged_df['DESCRIPTION_med'].apply(extract_dosage_numeric)
if 'START_med' in merged_df.columns and 'STOP_med' in merged_df.columns:
    valid_duration_mask = merged_df['START_med'].notna() & merged_df['STOP_med'].notna()
    merged_df['trialDuration'] = np.nan
    merged_df.loc[valid_duration_mask, 'trialDuration'] = (merged_df.loc[valid_duration_mask, 'STOP_med'] - merged_df.loc[valid_duration_mask, 'START_med']).dt.days
    merged_df['trialDuration'] = merged_df['trialDuration'].clip(lower=1)
else: merged_df['trialDuration'] = np.nan

# Temporal Aggregation
SIDE_EFFECT_KEYWORDS = ['reaction', 'adverse', 'side effect', 'rash', 'nausea', 'vomiting', 'dizziness', 'headache', 'allergic', 'pain', 'swelling', 'abnormal', 'elevated', 'decreased', 'bleed', 'infection']
IMPROVEMENT_KEYWORDS = ['improvement', 'improved', 'better', 'resolved', 'stable', 'normal', 'effective', 'relief', 'decreased', 'controlled']
WORSENING_KEYWORDS = ['worsening', 'worsened', 'decline', 'increased', 'no improvement', 'failed', 'exacerbation', 'progression', 'severe']
def aggregate_temporal_data(group):
    med_start = group['START_med'].iloc[0] if 'START_med' in group.columns and not group['START_med'].empty else pd.NaT
    med_stop = group['STOP_med'].iloc[0] if 'STOP_med' in group.columns and not group['STOP_med'].empty else pd.NaT
    results = {
        'conditions_during_encounter': [], 'new_conditions_after_med_start': [], 'trialSideEffects': [],
        'aggregated_observations': {}, 'count_side_effect_keywords_obs': 0, 'count_improvement_keywords_obs': 0,
        'count_worsening_keywords_obs': 0, 'has_impossible_observation': False,
        'avg_temp_during_trial': np.nan, 'max_hr_during_trial': np.nan,
        'min_bp_systolic_during_trial': np.nan, 'max_bp_systolic_during_trial': np.nan,
        'min_bp_diastolic_during_trial': np.nan, 'max_bp_diastolic_during_trial': np.nan,
    }
    if pd.isna(med_start) or pd.isna(med_stop): return pd.Series(results)
    conditions_during = set(); new_conditions = set()
    if 'DESCRIPTION_cond' in group.columns:
        relevant_conditions = group[group['DESCRIPTION_cond'].notna() & group['START_cond'].notna() & (((group['START_cond'] <= med_stop) & (group['STOP_cond'].isna() | (group['STOP_cond'] >= med_start))) | ((group['START_cond'] >= med_start) & (group['START_cond'] <= med_stop)))].drop_duplicates(subset=['DESCRIPTION_cond'])
        for _, row in relevant_conditions.iterrows():
            desc = str(row['DESCRIPTION_cond']).strip().title(); conditions_during.add(desc)
            if row['START_cond'] > med_start: new_conditions.add(desc)
    results['conditions_during_encounter'] = list(conditions_during); results['new_conditions_after_med_start'] = list(new_conditions)
    temps, hrs, bps = [], [], []
    if 'OBS_DATE_obs' in group.columns:
        relevant_obs = group[group['OBS_DATE_obs'].notna() & (group['OBS_DATE_obs'] >= med_start) & (group['OBS_DATE_obs'] <= med_stop)].dropna(subset=['DESCRIPTION_obs', 'VALUE_obs'])
        for _, row in relevant_obs.iterrows():
            desc, val_str, unit, code = str(row['DESCRIPTION_obs']).lower(), str(row['VALUE_obs']).strip(), str(row['UNITS_obs']).lower(), str(row['CODE_obs'])
            val_num = pd.to_numeric(val_str, errors='coerce')
            if any(keyword in desc for keyword in SIDE_EFFECT_KEYWORDS): results['count_side_effect_keywords_obs'] += 1
            if any(keyword in desc for keyword in IMPROVEMENT_KEYWORDS): results['count_improvement_keywords_obs'] += 1
            if any(keyword in desc for keyword in WORSENING_KEYWORDS): results['count_worsening_keywords_obs'] += 1
            if not pd.isna(val_num):
                if 'temperature' in desc or code == '8310-5':
                    temp_c = val_num if 'c' in unit else ((val_num - 32) * 5/9 if 'f' in unit else np.nan)
                    if 25 < temp_c < 45: temps.append(temp_c)
                    elif temp_c < 20 or temp_c > 50: results['has_impossible_observation'] = True
                elif 'heart rate' in desc or code == '8867-4':
                    if 20 < val_num < 250: hrs.append(val_num)
                    elif val_num < 10 or val_num > 300: results['has_impossible_observation'] = True
                elif 'blood pressure' in desc or code in ['85354-9', '55284-4']:
                    bp_match = re.match(r'(\d+)\s*/\s*(\d+)', val_str)
                    if bp_match:
                        systolic, diastolic = map(float, bp_match.groups())
                        if 30 < systolic < 300 and 20 < diastolic < 200: bps.append((systolic, diastolic))
                        else: results['has_impossible_observation'] = True
                    elif code == '8480-6': # Systolic
                         if 30 < val_num < 300: bps.append((val_num, np.nan))
                         else: results['has_impossible_observation'] = True
                    elif code == '8462-4': # Diastolic
                         if 20 < val_num < 200: bps.append((np.nan, val_num))
                         else: results['has_impossible_observation'] = True
        if temps: results['avg_temp_during_trial'] = np.mean(temps)
        if hrs: results['max_hr_during_trial'] = np.max(hrs)
        if bps:
            valid_systolics = [s for s, d in bps if not pd.isna(s)]; valid_diastolics = [d for s, d in bps if not pd.isna(d)]
            if valid_systolics: results['min_bp_systolic_during_trial'], results['max_bp_systolic_during_trial'] = np.min(valid_systolics), np.max(valid_systolics)
            if valid_diastolics: results['min_bp_diastolic_during_trial'], results['max_bp_diastolic_during_trial'] = np.min(valid_diastolics), np.max(valid_diastolics)
    side_effect_signals = set(results['new_conditions_after_med_start'])
    if results['count_side_effect_keywords_obs'] > 0: side_effect_signals.add("Observation Keyword Trigger")
    results['trialSideEffects'] = list(side_effect_signals)
    # We don't need the intermediate 'aggregated_observations' dict here
    results.pop('aggregated_observations', None) # Remove intermediate dict
    return results # Return dictionary

grouping_cols = ['PATIENT', 'ENCOUNTER', 'START_med']
if all(col in merged_df.columns for col in grouping_cols):
    merged_df.dropna(subset=grouping_cols, inplace=True)
    if not merged_df[grouping_cols].isnull().any().any():
        grouped = merged_df.groupby(grouping_cols, observed=True, group_keys=True) # group_keys=True might help
        aggregated_results_list = []
        print("Iterating through groups for aggregation...")
        for name, group_df in grouped:
            # name is a tuple of the grouping keys (PATIENT, ENCOUNTER, START_med)
            agg_result_dict = aggregate_temporal_data(group_df)
            # Add grouping keys back to the result dictionary
            agg_result_dict['PATIENT'] = name[0]
            agg_result_dict['ENCOUNTER'] = name[1]
            agg_result_dict['START_med'] = name[2]
            aggregated_results_list.append(agg_result_dict)

        if not aggregated_results_list:
             print("Warning: No groups found or processed during aggregation.")
             # Handle empty aggregation result - maybe create empty merged_df?
             merged_df = pd.DataFrame(columns=original_cols_needed + grouping_cols + list(aggregated_results_list[0].keys() if aggregated_results_list else [])) # Placeholder structure
        else:
            aggregated_features_df = pd.DataFrame(aggregated_results_list)
            print(f"DEBUG: Columns in aggregated_features_df from list: {aggregated_features_df.columns.tolist()}")

            # Explicitly define original columns needed besides grouping keys
            original_cols_needed = [
            'gender', 'age', 'doctorNotes', 'noteLength', 'noteWordCount',
            'drugName', 'dosage_numeric', 'trialDuration', 'knownAllergies',
            'proc_count', 'claim_count'
        ]
        # Filter to columns actually present in merged_df before aggregation
        original_cols_to_keep = [col for col in original_cols_needed if col in merged_df.columns]
        # Ensure grouping columns are included for the merge key and drop duplicates
        original_data_subset = merged_df[grouping_cols + original_cols_to_keep].drop_duplicates(subset=grouping_cols)

        # Merge aggregated features back, explicitly specifying join keys
        merged_df = pd.merge(
            original_data_subset,
            aggregated_features_df, # Use the dataframe created above
            on=grouping_cols,       # Join keys are now columns in both
            how='left'
        )
        print(f"  Shape after merging aggregated features: {merged_df.shape}")

    else: print("Warning: NaNs found in grouping columns after dropna, skipping aggregation.")
else: print("Warning: Grouping columns missing, skipping temporal aggregation.")

# Derive Placeholder Features
def map_severity(row):
    side_effects = row.get('trialSideEffects', [])
    worsening_count = row.get('count_worsening_keywords_obs', 0)
    if not isinstance(side_effects, list): side_effects = []
    if 'Observation Keyword Trigger' in side_effects and worsening_count > 0: return 'Severe'
    if len(side_effects) > 1: return 'Moderate'
    if len(side_effects) == 1: return 'Mild'
    return 'None'
merged_df['sideEffectSeverity'] = merged_df.apply(map_severity, axis=1) if 'trialSideEffects' in merged_df.columns else 'None'
def map_status(row):
    worsening_count = row.get('count_worsening_keywords_obs', 0)
    improvement_count = row.get('count_improvement_keywords_obs', 0)
    new_conditions = row.get('new_conditions_after_med_start', [])
    if not isinstance(new_conditions, list): new_conditions = []
    if worsening_count > 0: return 'Worse'
    if improvement_count > 0: return 'Improved'
    if len(new_conditions) > 0: return 'Worse'
    return 'Same'
merged_df['overallHealthStatus'] = merged_df.apply(map_status, axis=1) if 'count_worsening_keywords_obs' in merged_df.columns else 'Same'
def map_improvement(row):
    improvement_count = row.get('count_improvement_keywords_obs', 0)
    worsening_count = row.get('count_worsening_keywords_obs', 0)
    if improvement_count > worsening_count: return 7
    if worsening_count > improvement_count: return 3
    return 5
merged_df['symptomImprovementScore'] = merged_df.apply(map_improvement, axis=1) if 'count_improvement_keywords_obs' in merged_df.columns else 5

# Drop intermediate columns
cols_to_drop_now = ['birthdate', 'DATE', 'DESCRIPTION', 'REASONDESCRIPTION', 'DESCRIPTION_med', 'START_med', 'STOP_med', 'PATIENT_cond_ref', 'ENCOUNTER_cond_ref', 'DESCRIPTION_cond', 'START_cond', 'STOP_cond', 'PATIENT_obs_ref', 'ENCOUNTER_obs_ref', 'OBS_DATE_obs', 'CODE_obs', 'VALUE_obs', 'UNITS_obs', 'DESCRIPTION_obs', 'observedSideEffects']
merged_df = merged_df.drop(columns=cols_to_drop_now, errors='ignore')

# Drop duplicates
key_cols = ['PATIENT', 'ENCOUNTER', 'drugName']
if all(k in merged_df.columns for k in key_cols):
    merged_df = merged_df.drop_duplicates(subset=key_cols, keep='first')

df_processed = merged_df
del merged_df

if df_processed.empty:
    print("Error: Dataframe is empty after preprocessing. Cannot proceed.")
    exit()

# Post-Aggregation Imputation & Final Calculations (Dosage/Drug Freq)
num_cols_to_impute = ['age', 'dosage_numeric', 'trialDuration', 'avg_temp_during_trial', 'max_hr_during_trial', 'min_bp_systolic_during_trial', 'max_bp_systolic_during_trial', 'min_bp_diastolic_during_trial', 'max_bp_diastolic_during_trial', 'count_side_effect_keywords_obs', 'count_improvement_keywords_obs', 'count_worsening_keywords_obs']
for col in num_cols_to_impute:
    if col in df_processed.columns: df_processed[col] = df_processed[col].fillna(df_processed[col].median())
if 'has_impossible_observation' in df_processed.columns: df_processed['has_impossible_observation'] = df_processed['has_impossible_observation'].fillna(False)
list_cols = ['knownAllergies', 'conditions_during_encounter', 'new_conditions_after_med_start', 'trialSideEffects']
for col in list_cols:
    if col in df_processed.columns: df_processed[col] = df_processed[col].apply(lambda x: x if isinstance(x, list) else [])
drug_freq_map = {}; dosage_stats = {}
if 'drugName' in df_processed.columns:
    drug_freq = df_processed['drugName'].value_counts(normalize=True); df_processed['drugName_freq'] = df_processed['drugName'].map(drug_freq).fillna(0); drug_freq_map = drug_freq.to_dict()
else: df_processed['drugName_freq'] = 0
if 'dosage_numeric' in df_processed.columns:
    dosage_stats['overall_mean'] = df_processed['dosage_numeric'].mean(); dosage_stats['overall_std'] = df_processed['dosage_numeric'].std(); dosage_stats['overall_p01'] = df_processed['dosage_numeric'].quantile(0.01); dosage_stats['overall_p99'] = df_processed['dosage_numeric'].quantile(0.99)
    mean, std, p01, p99 = dosage_stats['overall_mean'], dosage_stats['overall_std'], dosage_stats['overall_p01'], dosage_stats['overall_p99']
    df_processed['dosage_zscore_overall'] = (df_processed['dosage_numeric'] - mean) / std if std != 0 else 0
    df_processed['isDosageOutOfRange_overall'] = ~df_processed['dosage_numeric'].between(p01, p99, inclusive='both')
else:
    dosage_stats = {'overall_mean': 0, 'overall_std': 1, 'overall_p01': 0, 'overall_p99': 1}; df_processed['dosage_zscore_overall'] = 0; df_processed['isDosageOutOfRange_overall'] = False
if 'isDosageOutOfRange_overall' in df_processed.columns: df_processed['isDosageOutOfRange_overall'] = df_processed['isDosageOutOfRange_overall'].fillna(False)

# Split data (using the same split size and random state)
train_df, _ = train_test_split(df_processed, test_size=VALIDATION_SPLIT_SIZE, random_state=42)
print(f"Training data shape for preprocessor fitting: {train_df.shape}")

# Define Feature Groups (ensure consistency)
numerical_features = ['age', 'trialDuration', 'dosage_numeric', 'drugName_freq', 'noteLength', 'noteWordCount', 'proc_count', 'claim_count', 'dosage_zscore_overall', 'avg_temp_during_trial', 'max_hr_during_trial', 'min_bp_systolic_during_trial', 'max_bp_systolic_during_trial', 'min_bp_diastolic_during_trial', 'max_bp_diastolic_during_trial', 'count_side_effect_keywords_obs', 'count_improvement_keywords_obs', 'count_worsening_keywords_obs']
categorical_features = ['gender', 'sideEffectSeverity', 'overallHealthStatus', 'symptomImprovementScore', 'isDosageOutOfRange_overall', 'has_impossible_observation']
text_feature = 'doctorNotes'
multilabel_features = ['knownAllergies', 'conditions_during_encounter', 'new_conditions_after_med_start', 'trialSideEffects']

# Load MLB classes from JSON or derive from training data
try:
    with open(OPTIONS_JSON_FILE, 'r') as f: frontend_options = json.load(f)
    all_known_allergies_classes = frontend_options.get('knownAllergies', [])
    all_conditions_during_classes = frontend_options.get('conditions_during_encounter', []) # Key might differ, adjust if needed
    all_new_conditions_classes = frontend_options.get('new_conditions_after_med_start', [])
    all_trial_sideeffects_classes = frontend_options.get('trialSideEffects', []) # Key might differ, adjust if needed
    def get_classes_from_data(df, col_name):
        if col_name not in df.columns: return []
        all_items = set(); df[col_name].apply(lambda lst: all_items.update(lst)); return sorted(list(all_items))
    if not all_known_allergies_classes: all_known_allergies_classes = get_classes_from_data(train_df, 'knownAllergies')
    if not all_conditions_during_classes: all_conditions_during_classes = get_classes_from_data(train_df, 'conditions_during_encounter')
    if not all_new_conditions_classes: all_new_conditions_classes = get_classes_from_data(train_df, 'new_conditions_after_med_start')
    if not all_trial_sideeffects_classes: all_trial_sideeffects_classes = get_classes_from_data(train_df, 'trialSideEffects')
except Exception as e: print(f"Warning: Error loading options for MLB: {e}"); all_known_allergies_classes, all_conditions_during_classes, all_new_conditions_classes, all_trial_sideeffects_classes = [], [], [], []
multilabel_classes = {'knownAllergies': all_known_allergies_classes, 'conditions_during_encounter': all_conditions_during_classes, 'new_conditions_after_med_start': all_new_conditions_classes, 'trialSideEffects': all_trial_sideeffects_classes}

# Verify features exist in train_df
train_cols = train_df.columns.tolist()
numerical_features = [f for f in numerical_features if f in train_cols]
categorical_features = [f for f in categorical_features if f in train_cols]
multilabel_features = [f for f in multilabel_features if f in train_cols]
if text_feature and text_feature not in train_cols: text_feature = None

# Create and Fit Preprocessing Pipeline ONCE
numeric_transformer = Pipeline([('imputer', SimpleImputer(strategy='median')), ('scaler', StandardScaler())])
categorical_transformer = Pipeline([('imputer', SimpleImputer(strategy='most_frequent')), ('onehot', OneHotEncoder(handle_unknown='ignore', sparse_output=False))])
# Create a text processing pipeline that cleans THEN vectorizes
text_pipeline = Pipeline([
    ('cleaner', FunctionTransformer(clean_text_series, validate=False)), # Apply cleaning function
    ('vectorizer', TfidfVectorizer(max_features=TFIDF_MAX_FEATURES, stop_words='english')) # Then vectorize
])
transformers_list = []
if numerical_features: transformers_list.append(('num', numeric_transformer, numerical_features))
if categorical_features: transformers_list.append(('cat', categorical_transformer, categorical_features))
# Use the text_pipeline for the text feature
if text_feature: transformers_list.append(('txt', text_pipeline, text_feature))
if not transformers_list: print("Error: No features available for ColumnTransformer."); exit()
ct = ColumnTransformer(transformers=transformers_list, remainder='drop')
print("Fitting ColumnTransformer...")
ct.fit(train_df)
print("ColumnTransformer fitting complete.")
mlb_transformers = {}
print("Fitting MLBs...")
for col in multilabel_features:
    classes = multilabel_classes.get(col, [])
    if classes and col in train_df.columns:
        mlb = MultiLabelBinarizer(classes=classes); mlb.fit([classes]); mlb_transformers[col] = mlb
print("MLB fitting complete.")

# Combine fitted preprocessor components (needed for transforming test data later)
preprocessor_artifact = { 'col_transformer': ct, 'mlb_transformers': mlb_transformers, 'mlb_classes': multilabel_classes }

# Fit the preprocessor on the training data to get X_train_processed for model training
print("Preprocessing training data for model fitting...")
X_train_ct = ct.transform(train_df)
if hasattr(X_train_ct, "toarray"): X_train_ct = X_train_ct.toarray()
X_train_mlb_features = []
for col in multilabel_features:
    if col in mlb_transformers:
        mlb = mlb_transformers[col]
        transformed_data = mlb.transform(train_df[col].apply(lambda x: [item for item in x if item in mlb.classes_]))
        X_train_mlb_features.append(transformed_data)
    else:
        classes = multilabel_classes.get(col, [])
        X_train_mlb_features.append(np.zeros((len(train_df), len(classes))))
X_train_mlb_features_dense = [arr.toarray() if hasattr(arr, "toarray") else arr for arr in X_train_mlb_features]
X_train_processed = np.hstack([X_train_ct] + X_train_mlb_features_dense)
print(f"Training data preprocessed. Shape: {X_train_processed.shape}")
print("--- Step 1 Complete ---")


# --- 2. Prepare Test Data ---
print("\n--- Step 2: Preparing Test Data ---")
test_data_dicts = [tc['data'] for tc in test_cases]
test_df = pd.DataFrame(test_data_dicts)

# Apply the same feature engineering steps to test_df as needed for prediction
# (Dosage/Drug Freq calculation uses maps/stats from training data)
test_df['dosage_numeric'] = pd.to_numeric(test_df['dosage_numeric'], errors='coerce')
test_df['drugName'] = test_df['drugName'].astype(str)
# Use maps/stats derived from the full processed training data before split
test_df['drugName_freq'] = test_df['drugName'].map(drug_freq_map).fillna(0)
mean, std, p01, p99 = dosage_stats['overall_mean'], dosage_stats['overall_std'], dosage_stats['overall_p01'], dosage_stats['overall_p99']
test_df['dosage_numeric'] = test_df['dosage_numeric'].fillna(mean) # Impute test data NaNs
test_df['dosage_zscore_overall'] = (test_df['dosage_numeric'] - mean) / std if std != 0 else 0
test_df['isDosageOutOfRange_overall'] = ~test_df['dosage_numeric'].between(p01, p99, inclusive='both')
test_df['isDosageOutOfRange_overall'] = test_df['isDosageOutOfRange_overall'].fillna(False) # Impute flag NaNs

# Ensure list columns are lists
for col in multilabel_features:
    if col in test_df.columns:
        test_df[col] = test_df[col].apply(lambda x: parse_list_string(x) if not isinstance(x, list) else x)
    else:
        test_df[col] = [[] for _ in range(len(test_df))]

# Add missing columns with defaults if necessary (based on fitted preprocessor)
ct_feature_names = []
try: # Get feature names expected by CT
    for name, _, columns in ct.transformers_:
        if name != 'remainder': ct_feature_names.extend(columns)
except AttributeError: # Handle older sklearn versions if needed
     ct_feature_names = numerical_features + categorical_features + ([text_feature] if text_feature else [])

all_expected_features = ct_feature_names + multilabel_features
for feature in all_expected_features:
    if feature not in test_df.columns:
        default_value = 0 if feature in numerical_features else ('Unknown' if feature in categorical_features else ('' if feature == text_feature else ([] if feature in multilabel_features else None)))
        test_df[feature] = default_value
        print(f"Warning: Added missing column '{feature}' to test data with default.")

# Preprocess test data using the FITTED preprocessor
print("Preprocessing test data...")
try:
    # Removed explicit pre-cleaning here, as it's now handled inside the text_pipeline
    X_test_ct = ct.transform(test_df) # Use transform, not fit_transform
    if hasattr(X_test_ct, "toarray"): X_test_ct = X_test_ct.toarray()

    X_test_mlb_features = []
    for col in multilabel_features:
        if col in mlb_transformers:
            mlb = mlb_transformers[col]
            # Filter test data items to known classes before transform
            filtered_test_col_data = test_df[col].apply(lambda x: [item for item in x if item in mlb.classes_])
            transformed_data = mlb.transform(filtered_test_col_data)
            X_test_mlb_features.append(transformed_data)
        else:
            classes = multilabel_classes.get(col, [])
            X_test_mlb_features.append(np.zeros((len(test_df), len(classes))))

    X_test_mlb_features_dense = [arr.toarray() if hasattr(arr, "toarray") else arr for arr in X_test_mlb_features]
    X_test_processed = np.hstack([X_test_ct] + X_test_mlb_features_dense)
    print(f"Test data preprocessed. Shape: {X_test_processed.shape}")
    test_data_ready = True
except Exception as e:
    print(f"Error preprocessing test data: {e}. Cannot proceed with tuning evaluation.")
    X_test_processed = None
    test_data_ready = False

print("--- Step 2 Complete ---")


# --- 3. Hyperparameter Tuning Loop ---
print("\n--- Step 3: Running Hyperparameter Tuning ---")
results = []

if test_data_ready:
    # --- LOF Tuning ---
    print("\nTuning LocalOutlierFactor...")
    for cont in LOF_CONTAMINATION_VALUES:
        print(f"  Training LOF with contamination={cont}...")
        try:
            lof = LocalOutlierFactor(n_neighbors=20, contamination=cont, novelty=True, n_jobs=-1)
            lof.fit(X_train_processed)
            print("    Predicting on test set...")
            y_pred_lof = lof.predict(X_test_processed) # Predict returns -1 for outliers, 1 for inliers

            # Evaluate
            print("    Evaluating...")
            accuracy = accuracy_score(true_labels, y_pred_lof)
            f1_anomaly = f1_score(true_labels, y_pred_lof, pos_label=-1, average='binary', zero_division=0)
            report = classification_report(true_labels, y_pred_lof, labels=[-1, 1], target_names=['Anomaly', 'Normal'], output_dict=True, zero_division=0)
            cm = confusion_matrix(true_labels, y_pred_lof, labels=[-1, 1])

            results.append({
                'model': 'LocalOutlierFactor',
                'params': {'contamination': cont},
                'accuracy': accuracy,
                'f1_anomaly': f1_anomaly,
                'report': report,
                'confusion_matrix': cm.tolist() # Convert numpy array to list for printing/storage
            })
            print(f"    Done. Accuracy: {accuracy:.4f}, F1 (Anomaly): {f1_anomaly:.4f}")
        except Exception as e:
            print(f"    Error training/evaluating LOF with contamination={cont}: {e}")
            results.append({
                'model': 'LocalOutlierFactor',
                'params': {'contamination': cont},
                'error': str(e)
            })

    # --- OCSVM Tuning ---
    print("\nTuning OneClassSVM...")
    for n in OCSVM_NU_VALUES:
        for g in OCSVM_GAMMA_VALUES:
            print(f"  Training OCSVM with nu={n}, gamma={g}...")
            try:
                ocsvm = OneClassSVM(gamma=g, nu=n)
                ocsvm.fit(X_train_processed)
                print("    Predicting on test set...")
                y_pred_ocsvm = ocsvm.predict(X_test_processed) # Predict returns -1 for outliers, 1 for inliers

                # Evaluate (Ensure this is inside the try block)
                print("    Evaluating...")
                accuracy = accuracy_score(true_labels, y_pred_ocsvm)
                f1_anomaly = f1_score(true_labels, y_pred_ocsvm, pos_label=-1, average='binary', zero_division=0)
                report = classification_report(true_labels, y_pred_ocsvm, labels=[-1, 1], target_names=['Anomaly', 'Normal'], output_dict=True, zero_division=0)
                cm = confusion_matrix(true_labels, y_pred_ocsvm, labels=[-1, 1])

                results.append({
                    'model': 'OneClassSVM',
                    'params': {'nu': n, 'gamma': g}, # Store both params
                    'accuracy': accuracy,
                    'f1_anomaly': f1_anomaly,
                    'report': report,
                    'confusion_matrix': cm.tolist()
                })
                print(f"    Done. Accuracy: {accuracy:.4f}, F1 (Anomaly): {f1_anomaly:.4f}")
            except Exception as e:
                print(f"    Error training/evaluating OCSVM with nu={n}, gamma={g}: {e}")
                results.append({
                    'model': 'OneClassSVM',
                    'params': {'nu': n, 'gamma': g},
                    'error': str(e)
                })
else:
    print("Skipping tuning loop due to errors in test data preparation.")

print("--- Step 3 Complete ---")


# --- 4. Analyze Results and Select Best ---
print("\n--- Step 4: Analyzing Tuning Results ---")

best_lof = None
best_ocsvm = None
best_lof_score = -1
best_ocsvm_score = -1

print("\nSummary of Results:")
print("-" * 60)
print("| Model              | Parameters         | Accuracy | F1 (Anomaly) |")
print("|--------------------|--------------------|----------|--------------|")

for res in results:
    if 'error' in res:
        print(f"| {res['model']:<18} | {str(res['params']):<18} | Error    | Error        |")
    else:
        model_name = res['model']
        params_str = str(res['params'])
        acc = res['accuracy']
        f1 = res['f1_anomaly']
        print(f"| {model_name:<18} | {params_str:<18} | {acc:<8.4f} | {f1:<12.4f} |")

        # Select best based on F1 score for the anomaly class (-1)
        if model_name == 'LocalOutlierFactor' and f1 > best_lof_score:
            best_lof_score = f1
            best_lof = res
        elif model_name == 'OneClassSVM' and f1 > best_ocsvm_score:
            best_ocsvm_score = f1
            best_ocsvm = res

print("-" * 60)

print("\nBest Parameters Found (based on F1 score for Anomaly class):")
if best_lof:
    print(f"\nLocalOutlierFactor:")
    print(f"  Parameters: {best_lof['params']}")
    print(f"  Accuracy: {best_lof['accuracy']:.4f}")
    print(f"  F1 (Anomaly): {best_lof['f1_anomaly']:.4f}")
    print("  Classification Report:")
    print(classification_report(true_labels, # Need to re-predict or store predictions if needed
                                LocalOutlierFactor(**best_lof['params'], novelty=True, n_jobs=-1).fit(X_train_processed).predict(X_test_processed),
                                labels=[-1, 1], target_names=['Anomaly', 'Normal'], zero_division=0))
    print("  Confusion Matrix:")
    print(best_lof['confusion_matrix'])
else:
    print("\nLocalOutlierFactor: No successful runs or best model found.")

if best_ocsvm:
    print(f"\nOneClassSVM:")
    print(f"  Parameters: {best_ocsvm['params']}")
    print(f"  Accuracy: {best_ocsvm['accuracy']:.4f}")
    print(f"  F1 (Anomaly): {best_ocsvm['f1_anomaly']:.4f}")
    print("  Classification Report:")
    # Re-predict using only the best parameters found during tuning
    print(classification_report(true_labels,
                                OneClassSVM(**best_ocsvm['params']).fit(X_train_processed).predict(X_test_processed),
                                labels=[-1, 1], target_names=['Anomaly', 'Normal'], zero_division=0))
    print("  Confusion Matrix:")
    print(best_ocsvm['confusion_matrix'])
else:
    print("\nOneClassSVM: No successful runs or best model found.")

print("\n--- Step 4 Complete ---")
print("\nHyperparameter tuning script finished.")
```

```python
import pandas as pd
import numpy as np
import os
import json
import joblib
import re
import warnings
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, OneHotEncoder, MultiLabelBinarizer
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.ensemble import IsolationForest
from sklearn.neighbors import LocalOutlierFactor
from sklearn.svm import OneClassSVM
from sklearn.pipeline import Pipeline
import spacy
from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer
import tensorflow as tf
from tensorflow import keras
from keras import layers, losses

# Suppress warnings and TensorFlow logs
warnings.filterwarnings('ignore')
os.environ['TF_CPP_MIN_LOG_LEVEL'] = '2' # Suppress TensorFlow INFO messages
tf.get_logger().setLevel('ERROR') # Suppress TensorFlow warnings

# --- Load spaCy Model ---
NLP = None
try:
    # Note: Requires spaCy model to be downloaded first!
    # Run: python -m spacy download en_core_web_sm
    NLP = spacy.load("en_core_web_sm")
    print("spaCy model 'en_core_web_sm' loaded successfully.")
except OSError:
    print("Warning: spaCy model 'en_core_web_sm' not found. ")
    print("NLP features will not be generated.")
    print("Download it by running: python -m spacy download en_core_web_sm")

# --- Configuration ---
# Use environment variables with defaults
BASE_DIR = os.getenv('BASE_DIR', 'archive')
OPTIONS_JSON_FILE = os.getenv('OPTIONS_JSON_FILE', 'frontend_options.json') # For full list of MLB classes
# Default to a new tuned artifact file name
OUTPUT_ARTIFACTS_FILE = os.getenv('OUTPUT_ARTIFACTS_FILE', 'model_artifacts_tuned.joblib')
VALIDATION_SPLIT_SIZE = float(os.getenv('VALIDATION_SPLIT_SIZE', 0.2))
TFIDF_MAX_FEATURES = int(os.getenv('TFIDF_MAX_FEATURES', 1000))
# Tuned Hyperparameters
LOF_CONTAMINATION = os.getenv('LOF_CONTAMINATION', 'auto')
OCSVM_NU = float(os.getenv('OCSVM_NU', 0.1)) # Use best nu found from tuning

# Columns needed from each file
PATIENTS_COLS = ['patient', 'birthdate', 'gender']
ENCOUNTERS_COLS = ['ID', 'DATE', 'PATIENT', 'DESCRIPTION', 'REASONDESCRIPTION']
MEDICATIONS_COLS = ['PATIENT', 'ENCOUNTER', 'DESCRIPTION', 'START', 'STOP']
PROCEDURES_COLS = ['PATIENT', 'ENCOUNTER', 'DESCRIPTION']
ALLERGIES_COLS = ['PATIENT', 'DESCRIPTION']
CLAIMS_COLS = ['PATIENT', 'ENCOUNTER', 'DIAGNOSIS']
OBSERVATIONS_COLS = ['PATIENT', 'ENCOUNTER', 'DATE', 'CODE', 'VALUE', 'UNITS', 'DESCRIPTION']

# --- Helper Functions ---
def extract_dosage_numeric(desc):
    if pd.isna(desc): return np.nan
    match = re.search(r'(\d+(?:\.\d+)?)\s*(?:MG|MCG|ML)', str(desc), re.IGNORECASE)
    return float(match.group(1)) if match else np.nan

def extract_drug_name(desc):
     if pd.isna(desc): return "Unknown"
     pattern = re.compile(r'^([a-zA-Z\s-]+)(?:\s+\d|\s*\(|\s+MG|\s+MCG|\s+ML)', re.IGNORECASE)
     match = pattern.match(str(desc))
     if match:
         drug = match.group(1).strip().title()
         return drug if len(drug) > 3 else "Unknown"
     return "Unknown"

# --- Sentiment Analysis Function ---
SIA = SentimentIntensityAnalyzer()
def get_sentiment_score(text):
    """Calculates the VADER compound sentiment score."""
    if pd.isna(text):
        return 0.0 # Neutral score for missing text
    try:
        # Ensure text is string
        text = str(text)
        vs = SIA.polarity_scores(text)
        return vs['compound']
    except Exception:
         # Handle potential errors during analysis
         return 0.0

# --- NLP Feature Extraction Function ---
def extract_nlp_features(text):
    """Extracts simple NLP features using spaCy."""
    if NLP is None or pd.isna(text) or not text.strip():
        # Return default values if spaCy not loaded or text is empty/NaN
        return pd.Series({
            'nlp_verb_count': 0, 'nlp_noun_count': 0, 'nlp_adj_count': 0, 'nlp_ner_count': 0
        })
    doc = NLP(text)
    verbs = len([token for token in doc if token.pos_ == 'VERB'])
    nouns = len([token for token in doc if token.pos_ == 'NOUN'])
    adjs = len([token for token in doc if token.pos_ == 'ADJ'])
    ner = len(doc.ents) # Count named entities
    return pd.Series({
        'nlp_verb_count': verbs, 'nlp_noun_count': nouns, 'nlp_adj_count': adjs, 'nlp_ner_count': ner
    })

def safe_load_csv(path, usecols=None):
    """Loads CSV safely, handling encoding errors and missing files."""
    try:
        return pd.read_csv(path, usecols=usecols, low_memory=False)
    except UnicodeDecodeError:
        try:
            return pd.read_csv(path, usecols=usecols, encoding='latin1', low_memory=False)
        except Exception as e: print(f"  Error loading {path} (latin1): {e}"); return None
    except FileNotFoundError: print(f"  File not found: {path}"); return None
    except ValueError as e: print(f"  Column error loading {path}: {e}"); return None # Handle missing usecols
    except Exception as e: print(f"  Error loading {path}: {e}"); return None

# --- 1. Load Necessary Raw Data Columns ---
print("Loading necessary raw data columns...")
df_patients = safe_load_csv(os.path.join(BASE_DIR, 'patients.csv'), PATIENTS_COLS)
df_encounters = safe_load_csv(os.path.join(BASE_DIR, 'encounters.csv'), ENCOUNTERS_COLS)
# Load conditions data
df_conditions = safe_load_csv(os.path.join(BASE_DIR, 'conditions.csv'), ['PATIENT', 'ENCOUNTER', 'DESCRIPTION', 'START', 'STOP'])
df_medications = safe_load_csv(os.path.join(BASE_DIR, 'medications.csv'), MEDICATIONS_COLS)
df_procedures = safe_load_csv(os.path.join(BASE_DIR, 'procedures.csv'), PROCEDURES_COLS)
df_allergies = safe_load_csv(os.path.join(BASE_DIR, 'allergies.csv'), ALLERGIES_COLS)
df_claims = safe_load_csv(os.path.join(BASE_DIR, 'claims.csv'), CLAIMS_COLS)
df_observations = safe_load_csv(os.path.join(BASE_DIR, 'observations.csv'), OBSERVATIONS_COLS)

# Check essential dataframes
if df_patients is None or df_encounters is None:
    print("Error: Essential dataframes ('patients', 'encounters') could not be loaded. Exiting.")
    exit()

# Standardize IDs immediately
df_patients = df_patients.rename(columns={'patient': 'PATIENT'})
df_encounters = df_encounters.rename(columns={'ID': 'ENCOUNTER'})
if df_conditions is not None: df_conditions = df_conditions.rename(columns={'patient': 'PATIENT', 'Id': 'ENCOUNTER', 'ID': 'ENCOUNTER'}, errors='ignore')
if df_medications is not None: df_medications = df_medications.rename(columns={'patient': 'PATIENT', 'Id': 'ENCOUNTER', 'ID': 'ENCOUNTER'}, errors='ignore')
if df_procedures is not None: df_procedures = df_procedures.rename(columns={'patient': 'PATIENT', 'Id': 'ENCOUNTER', 'ID': 'ENCOUNTER'}, errors='ignore')
if df_allergies is not None: df_allergies = df_allergies.rename(columns={'patient': 'PATIENT'}, errors='ignore')
if df_claims is not None: df_claims = df_claims.rename(columns={'patient': 'PATIENT', 'Id': 'ENCOUNTER', 'ID': 'CLAIM_ID'}, errors='ignore') # Rename claims ID
if df_observations is not None: df_observations = df_observations.rename(columns={'patient': 'PATIENT', 'Id': 'ENCOUNTER', 'ID': 'ENCOUNTER', 'DATE': 'OBS_DATE'}, errors='ignore') # Rename obs date

# Calculate Condition Duration early if possible
if df_conditions is not None and 'START' in df_conditions.columns and 'STOP' in df_conditions.columns:
    df_conditions['START_cond_dt'] = pd.to_datetime(df_conditions['START'], errors='coerce')
    df_conditions['STOP_cond_dt'] = pd.to_datetime(df_conditions['STOP'], errors='coerce')
    valid_cond_duration_mask = df_conditions['START_cond_dt'].notna() & df_conditions['STOP_cond_dt'].notna()
    df_conditions['conditionDuration'] = np.nan
    df_conditions.loc[valid_cond_duration_mask, 'conditionDuration'] = (df_conditions.loc[valid_cond_duration_mask, 'STOP_cond_dt'] - df_conditions.loc[valid_cond_duration_mask, 'START_cond_dt']).dt.days
    df_conditions['conditionDuration'] = df_conditions['conditionDuration'].clip(lower=0) # Duration cannot be negative
    print("  Calculated 'conditionDuration'.")
else:
    if df_conditions is not None: df_conditions['conditionDuration'] = np.nan
    print("  Warning: Could not calculate 'conditionDuration' (missing START/STOP columns or data).")


# --- 2. Aggregate Allergies (Patient Level) ---
print("\nAggregating allergies...")
patient_allergies = pd.DataFrame({'PATIENT': [], 'knownAllergies': []}) # Renamed feature
if df_allergies is not None and 'PATIENT' in df_allergies.columns and 'DESCRIPTION' in df_allergies.columns:
    patient_allergies = df_allergies.dropna(subset=['DESCRIPTION']).groupby('PATIENT')['DESCRIPTION'].apply(lambda x: list(set(str(s).strip().title() for s in x if str(s).strip()))).reset_index()
    patient_allergies = patient_allergies.rename(columns={'DESCRIPTION': 'knownAllergies'}) # Renamed feature
    print(f"  Aggregated allergies (as knownAllergies) for {len(patient_allergies)} patients.")
else:
    print("  Could not aggregate allergies (missing data/columns).")

# --- 3. Aggregate Procedures & Claims Counts ---
print("\nAggregating procedure and claim counts...")
encounter_proc_counts = pd.DataFrame({'PATIENT': [], 'ENCOUNTER': [], 'proc_count': []})
if df_procedures is not None and 'PATIENT' in df_procedures.columns and 'ENCOUNTER' in df_procedures.columns:
    encounter_proc_counts = df_procedures.groupby(['PATIENT', 'ENCOUNTER']).size().reset_index(name='proc_count')
    print(f"  Aggregated procedure counts for {len(encounter_proc_counts)} encounters.")
else:
    print("  Could not aggregate procedures (missing data/columns).")

encounter_claim_counts = pd.DataFrame({'PATIENT': [], 'ENCOUNTER': [], 'claim_count': []})
if df_claims is not None and 'PATIENT' in df_claims.columns and 'ENCOUNTER' in df_claims.columns:
    encounter_claim_counts = df_claims.groupby(['PATIENT', 'ENCOUNTER']).size().reset_index(name='claim_count')
    print(f"  Aggregated claim counts for {len(encounter_claim_counts)} encounters.")
else:
    print("  Could not aggregate claims (missing data/columns).")


# --- 4. Merge Data ---
print("\nMerging dataframes...")
merged_df = pd.merge(df_encounters, df_patients, on='PATIENT', how='left')
print(f"  Shape after patients: {merged_df.shape}")
# Prepare and merge medications, explicitly renaming columns
if df_medications is not None and 'PATIENT' in df_medications.columns and 'ENCOUNTER' in df_medications.columns:
    med_cols_to_merge = ['PATIENT', 'ENCOUNTER', 'DESCRIPTION', 'START', 'STOP']
    df_medications_renamed = df_medications[med_cols_to_merge].rename(columns={
        'DESCRIPTION': 'DESCRIPTION_med',
        'START': 'START_med',
        'STOP': 'STOP_med'
    })
    merged_df = pd.merge(merged_df, df_medications_renamed, on=['PATIENT', 'ENCOUNTER'], how='left')
    print(f"  Shape after medications: {merged_df.shape}")
else:
    merged_df['DESCRIPTION_med'] = pd.NA
    merged_df['START_med'] = pd.NaT
    merged_df['STOP_med'] = pd.NaT
    print("  Medications data not found or invalid, added placeholder columns.")

# Merge aggregated allergies (patient level)
merged_df = pd.merge(merged_df, patient_allergies, on='PATIENT', how='left')
print(f"  Shape after allergies: {merged_df.shape}")
# Merge aggregated counts (encounter level)
merged_df = pd.merge(merged_df, encounter_proc_counts, on=['PATIENT', 'ENCOUNTER'], how='left')
merged_df = pd.merge(merged_df, encounter_claim_counts, on=['PATIENT', 'ENCOUNTER'], how='left')
print(f"  Shape after counts: {merged_df.shape}")
# Merge raw conditions (encounter level) - include conditionDuration
if df_conditions is not None and 'PATIENT' in df_conditions.columns and 'ENCOUNTER' in df_conditions.columns:
    cond_cols_to_merge = ['PATIENT', 'ENCOUNTER', 'DESCRIPTION', 'START', 'STOP', 'conditionDuration'] # Add duration
    df_conditions_to_merge = df_conditions[[col for col in cond_cols_to_merge if col in df_conditions.columns]].rename(
        columns={'PATIENT': 'PATIENT_cond_ref', 'ENCOUNTER': 'ENCOUNTER_cond_ref', 'DESCRIPTION': 'DESCRIPTION_cond', 'START': 'START_cond', 'STOP': 'STOP_cond'}, errors='ignore'
    )
    merged_df = pd.merge(merged_df, df_conditions_to_merge, left_on=['PATIENT', 'ENCOUNTER'], right_on=['PATIENT_cond_ref', 'ENCOUNTER_cond_ref'], how='left')
    print(f"  Shape after conditions merge: {merged_df.shape}")
# Merge raw observations (encounter level)
if df_observations is not None and 'PATIENT' in df_observations.columns and 'ENCOUNTER' in df_observations.columns:
     df_observations_to_merge = df_observations.rename(columns={'PATIENT': 'PATIENT_obs_ref', 'ENCOUNTER': 'ENCOUNTER_obs_ref'}, errors='ignore')
     merged_df = pd.merge(merged_df, df_observations_to_merge, left_on=['PATIENT', 'ENCOUNTER'], right_on=['PATIENT_obs_ref', 'ENCOUNTER_obs_ref'], how='left')
     print(f"  Shape after observations merge: {merged_df.shape}")


# Fill NaN counts with 0
merged_df['proc_count'] = merged_df['proc_count'].fillna(0).astype(int)
merged_df['claim_count'] = merged_df['claim_count'].fillna(0).astype(int)
# Fill NaN lists with empty lists for allergies
merged_df['knownAllergies'] = merged_df['knownAllergies'].apply(lambda x: x if isinstance(x, list) else [])
# Initialize new list/dict columns needed for later feature engineering
merged_df['conditions_during_encounter'] = [[] for _ in range(len(merged_df))]
merged_df['new_conditions_after_med_start'] = [[] for _ in range(len(merged_df))]
merged_df['trialSideEffects'] = [[] for _ in range(len(merged_df))]
merged_df['aggregated_observations'] = [{} for _ in range(len(merged_df))]


# --- 5. Feature Engineering ---
print("\nPerforming feature engineering...")

# Convert date columns early on
print("  Converting date columns...")
merged_df['birthdate'] = pd.to_datetime(merged_df['birthdate'], errors='coerce')
merged_df['DATE'] = pd.to_datetime(merged_df['DATE'], errors='coerce') # Encounter date
merged_df['START_med'] = pd.to_datetime(merged_df['START_med'], errors='coerce') # Use renamed Medication start
merged_df['STOP_med'] = pd.to_datetime(merged_df['STOP_med'], errors='coerce') # Use renamed Medication stop
if 'OBS_DATE_obs' in merged_df.columns: merged_df['OBS_DATE_obs'] = pd.to_datetime(merged_df['OBS_DATE_obs'], errors='coerce')
if 'START_cond' in merged_df.columns: merged_df['START_cond'] = pd.to_datetime(merged_df['START_cond'], errors='coerce')
if 'STOP_cond' in merged_df.columns: merged_df['STOP_cond'] = pd.to_datetime(merged_df['STOP_cond'], errors='coerce')

# Age (calculated based on encounter date)
valid_dates_mask = merged_df['birthdate'].notna() & merged_df['DATE'].notna()
merged_df['age'] = np.nan
merged_df.loc[valid_dates_mask, 'age'] = (merged_df.loc[valid_dates_mask, 'DATE'] - merged_df.loc[valid_dates_mask, 'birthdate']).dt.days // 365.25
merged_df['age'] = merged_df['age'].clip(0, 120)
print("  Engineered 'age'.")

# Gender
if 'gender' in merged_df.columns:
    merged_df['gender'] = merged_df['gender'].map({'M': 'Male', 'F': 'Female'}).fillna('Other')
else: merged_df['gender'] = 'Other'
print("  Engineered 'gender'.")

# Doctor Notes & Text Features
# Ensure the correct columns from encounters are used (Note: pandas adds suffixes on merge conflicts)
encounter_desc_col = 'DESCRIPTION_x' # Corrected column name after merges
encounter_reason_col = 'REASONDESCRIPTION' # Assuming this is the correct original name
print(f"DEBUG: Columns right before creating doctorNotes: {merged_df.columns.tolist()}") # Moved debug print here
# Safer creation: convert to string, fillna, then combine
desc_str = merged_df[encounter_desc_col].astype(str).fillna('')
reason_str = merged_df[encounter_reason_col].astype(str).fillna('')
merged_df['doctorNotes'] = desc_str + ' | ' + reason_str
merged_df['doctorNotes'] = merged_df['doctorNotes'].replace(r'^\s*\|\s*$', '', regex=True).str.strip().fillna("No notes")
merged_df['noteLength'] = merged_df['doctorNotes'].apply(len)
merged_df['noteWordCount'] = merged_df['doctorNotes'].apply(lambda x: len(x.split()))
print("  Engineered 'doctorNotes', 'noteLength', 'noteWordCount'.")

# Apply NLP Feature Extraction
if NLP:
    print("  Applying NLP feature extraction...")
    nlp_features_df = merged_df['doctorNotes'].apply(extract_nlp_features)
    merged_df = pd.concat([merged_df, nlp_features_df], axis=1)
    print(f"    Added NLP features: {nlp_features_df.columns.tolist()}")
else:
    print("  Skipping NLP feature extraction (spaCy model not loaded).")
    # Add columns with default 0 if NLP failed
    merged_df['nlp_verb_count'] = 0
    merged_df['nlp_noun_count'] = 0
    merged_df['nlp_adj_count'] = 0
    merged_df['nlp_ner_count'] = 0

# Apply Sentiment Analysis
print("  Applying Sentiment Analysis...")
merged_df['sentiment_score'] = merged_df['doctorNotes'].apply(get_sentiment_score)
print("    Added 'sentiment_score' feature.")


# Drug Name & Dosage Numeric
merged_df['drugName'] = merged_df['DESCRIPTION_med'].apply(extract_drug_name)
merged_df['dosage_numeric'] = merged_df['DESCRIPTION_med'].apply(extract_dosage_numeric)
print("  Engineered 'drugName' and 'dosage_numeric'.")

# Dosage per Age (Moved after dosage_numeric is created)
# Handle potential division by zero or NaN age by replacing with NaN, then fillna
merged_df['dosage_per_age'] = (merged_df['dosage_numeric'] / merged_df['age'].replace(0, np.nan)).fillna(0)
print("  Engineered 'dosage_per_age'.")

# Trial Duration (using renamed columns)
if 'START_med' in merged_df.columns and 'STOP_med' in merged_df.columns:
    valid_duration_mask = merged_df['START_med'].notna() & merged_df['STOP_med'].notna()
    merged_df['trialDuration'] = np.nan
    merged_df.loc[valid_duration_mask, 'trialDuration'] = (merged_df.loc[valid_duration_mask, 'STOP_med'] - merged_df.loc[valid_duration_mask, 'START_med']).dt.days
    merged_df['trialDuration'] = merged_df['trialDuration'].clip(lower=1) # Min 1 day
    print("  Engineered 'trialDuration'.")
else:
    merged_df['trialDuration'] = np.nan
    print("  Warning: START_med or STOP_med columns missing, cannot calculate trialDuration.")


# --- Temporal Filtering and Aggregation of Observations/Conditions ---
print("  Performing temporal filtering and aggregation...")

# Define keywords
SIDE_EFFECT_KEYWORDS = ['reaction', 'adverse', 'side effect', 'rash', 'nausea', 'vomiting', 'dizziness', 'headache', 'allergic', 'pain', 'swelling', 'abnormal', 'elevated', 'decreased', 'bleed', 'infection']
IMPROVEMENT_KEYWORDS = ['improvement', 'improved', 'better', 'resolved', 'stable', 'normal', 'effective', 'relief', 'decreased', 'controlled']
WORSENING_KEYWORDS = ['worsening', 'worsened', 'decline', 'increased', 'no improvement', 'failed', 'exacerbation', 'progression', 'severe']

# Define aggregation function
def aggregate_temporal_data(group):
    med_start = group['START_med'].iloc[0] if 'START_med' in group.columns and not group['START_med'].empty else pd.NaT
    med_stop = group['STOP_med'].iloc[0] if 'STOP_med' in group.columns and not group['STOP_med'].empty else pd.NaT
    results = {
        'conditions_during_encounter': [], 'new_conditions_after_med_start': [], 'trialSideEffects': [],
        'aggregated_observations': {}, 'count_side_effect_keywords_obs': 0, 'count_improvement_keywords_obs': 0,
        'count_worsening_keywords_obs': 0, 'has_impossible_observation': False,
        'avg_temp_during_trial': np.nan, 'max_hr_during_trial': np.nan,
        'min_bp_systolic_during_trial': np.nan, 'max_bp_systolic_during_trial': np.nan,
        'min_bp_diastolic_during_trial': np.nan, 'max_bp_diastolic_during_trial': np.nan,
        # Add variance/std dev features
        'std_temp_during_trial': np.nan, 'std_hr_during_trial': np.nan,
        'std_bp_systolic_during_trial': np.nan, 'std_bp_diastolic_during_trial': np.nan,
        'avg_condition_duration': np.nan # Add average condition duration during trial
    }
    if pd.isna(med_start) or pd.isna(med_stop): return pd.Series(results) # Return early if no valid med dates

    conditions_during = set()
    new_conditions = set()
    if 'DESCRIPTION_cond' in group.columns:
        relevant_conditions = group[
            group['DESCRIPTION_cond'].notna() & group['START_cond'].notna() &
            (((group['START_cond'] <= med_stop) & (group['STOP_cond'].isna() | (group['STOP_cond'] >= med_start))) |
             ((group['START_cond'] >= med_start) & (group['START_cond'] <= med_stop)))
        ].drop_duplicates(subset=['DESCRIPTION_cond'])
        for _, row in relevant_conditions.iterrows():
            desc = str(row['DESCRIPTION_cond']).strip().title()
            conditions_during.add(desc)
            if row['START_cond'] > med_start: new_conditions.add(desc)
    results['conditions_during_encounter'] = list(conditions_during)
    results['new_conditions_after_med_start'] = list(new_conditions)
    # Calculate average condition duration for conditions active during the trial
    if 'conditionDuration' in relevant_conditions.columns:
        valid_durations = relevant_conditions['conditionDuration'].dropna()
        if not valid_durations.empty:
            results['avg_condition_duration'] = valid_durations.mean()

    temps, hrs, bps = [], [], []
    if 'OBS_DATE_obs' in group.columns:
        relevant_obs = group[
            group['OBS_DATE_obs'].notna() & (group['OBS_DATE_obs'] >= med_start) & (group['OBS_DATE_obs'] <= med_stop)
        ].dropna(subset=['DESCRIPTION_obs', 'VALUE_obs'])
        for _, row in relevant_obs.iterrows():
            desc, val_str, unit, code = str(row['DESCRIPTION_obs']).lower(), str(row['VALUE_obs']).strip(), str(row['UNITS_obs']).lower(), str(row['CODE_obs'])
            val_num = pd.to_numeric(val_str, errors='coerce')
            if any(keyword in desc for keyword in SIDE_EFFECT_KEYWORDS): results['count_side_effect_keywords_obs'] += 1
            if any(keyword in desc for keyword in IMPROVEMENT_KEYWORDS): results['count_improvement_keywords_obs'] += 1
            if any(keyword in desc for keyword in WORSENING_KEYWORDS): results['count_worsening_keywords_obs'] += 1
            if not pd.isna(val_num):
                if 'temperature' in desc or code == '8310-5':
                    temp_c = val_num if 'c' in unit else ((val_num - 32) * 5/9 if 'f' in unit else np.nan)
                    if 25 < temp_c < 45: temps.append(temp_c)
                    elif temp_c < 20 or temp_c > 50: results['has_impossible_observation'] = True
                elif 'heart rate' in desc or code == '8867-4':
                    if 20 < val_num < 250: hrs.append(val_num)
                    elif val_num < 10 or val_num > 300: results['has_impossible_observation'] = True
                elif 'blood pressure' in desc or code in ['85354-9', '55284-4']:
                    bp_match = re.match(r'(\d+)\s*/\s*(\d+)', val_str)
                    if bp_match:
                        systolic, diastolic = map(float, bp_match.groups())
                        if 30 < systolic < 300 and 20 < diastolic < 200: bps.append((systolic, diastolic))
                        else: results['has_impossible_observation'] = True
                    elif code == '8480-6': # Systolic
                         if 30 < val_num < 300: bps.append((val_num, np.nan))
                         else: results['has_impossible_observation'] = True
                    elif code == '8462-4': # Diastolic
                         if 20 < val_num < 200: bps.append((np.nan, val_num))
                         else: results['has_impossible_observation'] = True
        if temps: results['avg_temp_during_trial'] = np.mean(temps)
        if hrs: results['max_hr_during_trial'] = np.max(hrs)
        if bps:
            valid_systolics = [s for s, d in bps if not pd.isna(s)]
            valid_diastolics = [d for s, d in bps if not pd.isna(d)]
            if valid_systolics:
                results['min_bp_systolic_during_trial'] = np.min(valid_systolics)
                results['max_bp_systolic_during_trial'] = np.max(valid_systolics)
            if valid_diastolics:
                results['min_bp_diastolic_during_trial'] = np.min(valid_diastolics)
                results['max_bp_diastolic_during_trial'] = np.max(valid_diastolics)
                if len(valid_diastolics) > 1: results['std_bp_diastolic_during_trial'] = np.std(valid_diastolics)
        # Calculate std dev if more than one reading
        if len(temps) > 1: results['std_temp_during_trial'] = np.std(temps)
        if len(hrs) > 1: results['std_hr_during_trial'] = np.std(hrs)
        if len(valid_systolics) > 1: results['std_bp_systolic_during_trial'] = np.std(valid_systolics)

    side_effect_signals = set(results['new_conditions_after_med_start'])
    if results['count_side_effect_keywords_obs'] > 0: side_effect_signals.add("Observation Keyword Trigger")
    results['trialSideEffects'] = list(side_effect_signals)
    return pd.Series(results)

# Apply the aggregation function
grouping_cols = ['PATIENT', 'ENCOUNTER', 'START_med']
if not all(col in merged_df.columns for col in grouping_cols):
     print(f"Error: Grouping columns ({grouping_cols}) missing. Cannot perform temporal aggregation.")
     aggregated_features = None
else:
    # Drop rows where essential grouping columns are NaN *before* grouping
    initial_rows = len(merged_df)
    merged_df.dropna(subset=grouping_cols, inplace=True)
    if len(merged_df) < initial_rows:
        print(f"  Dropped {initial_rows - len(merged_df)} rows with NaN in grouping columns ({grouping_cols}).")

    # Check again after dropna
    if merged_df[grouping_cols].isnull().any().any():
        print(f"Error: Grouping columns ({grouping_cols}) still contain NaNs after dropna. Exiting.")
        exit()

    print(f"  Shape before aggregation grouping: {merged_df.shape}")
    grouped = merged_df.groupby(grouping_cols, observed=True, group_keys=False)
    aggregated_features = grouped.apply(aggregate_temporal_data)
    print(f"  Shape after aggregation: {aggregated_features.shape}")
    print(f"Columns in aggregated_features: {aggregated_features.columns.tolist()}")

    # Merge aggregated features back
    base_cols_to_keep = grouping_cols + [
        'DATE', 'REASONDESCRIPTION', 'birthdate', 'gender', 'DESCRIPTION_med', 'STOP_med',
        'knownAllergies', 'proc_count', 'claim_count', 'age', 'doctorNotes', 'noteLength',
        'noteWordCount', 'drugName', 'dosage_numeric', 'trialDuration',
        # Ensure NLP, Sentiment and Interaction features are kept
        'nlp_verb_count', 'nlp_noun_count', 'nlp_adj_count', 'nlp_ner_count', 'sentiment_score',
        'dosage_per_age'
    ]
    valid_base_cols = [col for col in base_cols_to_keep if col in grouped.first().columns]
    base_df = grouped.first()[valid_base_cols].reset_index()
    print(f"  Shape of base_df for merge: {base_df.shape}")
    print(f"  Columns in base_df: {base_df.columns.tolist()}")

    merged_df = pd.merge(base_df, aggregated_features, on=grouping_cols, how='left')
    print(f"  Shape after merging aggregated features: {merged_df.shape}")
    print(f"Columns AFTER aggregation merge: {merged_df.columns.tolist()}")

# --- Derive Placeholder Features ---
print("  Deriving placeholder features...")
# Placeholder for known severe side effects - needs domain expertise
KNOWN_SEVERE_SIDE_EFFECTS = {'Anaphylaxis', 'Kidney Failure', 'Severe Bleeding'} # Example

def map_severity(row):
    side_effects = row.get('trialSideEffects', [])
    worsening_count = row.get('count_worsening_keywords_obs', 0)
    if not isinstance(side_effects, list): side_effects = []

    has_known_severe = any(se in KNOWN_SEVERE_SIDE_EFFECTS for se in side_effects)

    # Refined logic: Known severe effect or keyword trigger + worsening count = Severe
    if has_known_severe or ('Observation Keyword Trigger' in side_effects and worsening_count > 0):
        return 'Severe'
    # Multiple side effects (excluding keyword trigger if no worsening) = Moderate
    if len([se for se in side_effects if se != "Observation Keyword Trigger"]) > 1:
         return 'Moderate'
    # Single specific side effect or keyword trigger alone = Mild
    if side_effects: # Check if list is not empty
         return 'Mild'
    # Otherwise None
    return 'None'
merged_df['sideEffectSeverity'] = merged_df.apply(map_severity, axis=1)

def map_status(row):
    worsening_count = row.get('count_worsening_keywords_obs', 0)
    improvement_count = row.get('count_improvement_keywords_obs', 0)
    new_conditions = row.get('new_conditions_after_med_start', [])
    if not isinstance(new_conditions, list): new_conditions = []
    if worsening_count > 0: return 'Worse'
    if improvement_count > 0: return 'Improved'
    if len(new_conditions) > 0: return 'Worse'
    return 'Same'
merged_df['overallHealthStatus'] = merged_df.apply(map_status, axis=1)

def map_improvement(row):
    improvement_count = row.get('count_improvement_keywords_obs', 0)
    worsening_count = row.get('count_worsening_keywords_obs', 0)
    if improvement_count > worsening_count: return 7
    if worsening_count > improvement_count: return 3
    return 5
merged_df['symptomImprovementScore'] = merged_df.apply(map_improvement, axis=1)
print("  Derived 'sideEffectSeverity', 'symptomImprovementScore', 'overallHealthStatus'.")


# Drop intermediate/raw columns used for engineering and aggregation helpers
cols_to_drop_now = [
    'birthdate', 'DATE', 'DESCRIPTION', 'REASONDESCRIPTION', 'DESCRIPTION_med',
    'START_med', 'STOP_med',
    'PATIENT_cond_ref', 'ENCOUNTER_cond_ref', 'DESCRIPTION_cond', 'START_cond', 'STOP_cond',
    'PATIENT_obs_ref', 'ENCOUNTER_obs_ref', 'OBS_DATE_obs', 'CODE_obs', 'VALUE_obs', 'UNITS_obs', 'DESCRIPTION_obs',
    'aggregated_observations' # Drop the intermediate dict
]
cols_to_drop_now.extend(['observedSideEffects']) # Drop old side effects if it exists
merged_df = merged_df.drop(columns=cols_to_drop_now, errors='ignore')
print(f"  Dropped intermediate columns. Final columns: {merged_df.columns.tolist()}")

# --- 6. Sampling ---
# Drop duplicates potentially introduced if aggregation key wasn't perfectly unique
key_cols = ['PATIENT', 'ENCOUNTER', 'drugName'] # Assuming drugName helps differentiate if multiple meds start same time
if all(k in merged_df.columns for k in key_cols):
    initial_rows = len(merged_df)
    merged_df = merged_df.drop_duplicates(subset=key_cols, keep='first')
    if len(merged_df) < initial_rows:
        print(f"  Shape after dropping duplicates on {key_cols}: {merged_df.shape}")

df_sampled = merged_df
print(f"Using full merged data. Shape before sampling/imputation: {df_sampled.shape}")
del merged_df

# --- 7. Post-Aggregation Imputation & Final Calculations ---
# Check if df_sampled is valid before proceeding
if not isinstance(df_sampled, pd.DataFrame) or df_sampled.empty:
    print("Error: Dataframe is empty or invalid after sampling step. Cannot proceed.")
    exit()

df_processed = df_sampled # Rename for clarity
# del df_sampled # Removed comment

print("\nPerforming post-aggregation imputation...")
# Impute NaNs in newly created numerical features from aggregation
num_cols_to_impute = [
    'age', 'dosage_numeric', 'trialDuration', #'dosage_zscore_overall', # Calculated later
    'avg_temp_during_trial', 'max_hr_during_trial',
    'min_bp_systolic_during_trial', 'max_bp_systolic_during_trial',
    'min_bp_diastolic_during_trial', 'max_bp_diastolic_during_trial',
    'count_side_effect_keywords_obs', 'count_improvement_keywords_obs', 'count_worsening_keywords_obs',
    # Add new features for imputation
    'avg_condition_duration', 'std_temp_during_trial', 'std_hr_during_trial',
    'std_bp_systolic_during_trial', 'std_bp_diastolic_during_trial',
    # Add NLP features for imputation
    'nlp_verb_count', 'nlp_noun_count', 'nlp_adj_count', 'nlp_ner_count',
    # Add sentiment score for imputation
    'sentiment_score',
    # Add interaction feature for imputation
    'dosage_per_age'
]
for col in num_cols_to_impute:
    if col in df_processed.columns:
        median_val = df_processed[col].median()
        df_processed[col] = df_processed[col].fillna(median_val)
        print(f"  Filled NaNs in '{col}' with median: {median_val:.2f}")
    else:
        print(f"  Warning: Column '{col}' not found for imputation.")

# Impute boolean feature (created during aggregation)
if 'has_impossible_observation' in df_processed.columns:
    df_processed['has_impossible_observation'] = df_processed['has_impossible_observation'].fillna(False)
    print("  Filled NaNs in 'has_impossible_observation' with False.")

# Ensure list features are lists (should be handled by initialization, but double-check)
list_cols = ['knownAllergies', 'conditions_during_encounter', 'new_conditions_after_med_start', 'trialSideEffects']
for col in list_cols:
    if col in df_processed.columns:
        df_processed[col] = df_processed[col].apply(lambda x: x if isinstance(x, list) else [])
    else:
          print(f"  Warning: List column '{col}' not found for checking.")

# --- Drug Frequency and Dosage Stats Calculation (on df_processed before split) ---
drug_freq_map = {}
if 'drugName' in df_processed.columns:
    drug_freq = df_processed['drugName'].value_counts(normalize=True)
    df_processed['drugName_freq'] = df_processed['drugName'].map(drug_freq).fillna(0)
    drug_freq_map = drug_freq.to_dict() # Convert to dict for saving
    print("  Engineered 'drugName_freq'.")
else:
    df_processed['drugName_freq'] = 0
    print("  Warning: 'drugName' column not found, 'drugName_freq' set to 0.")

dosage_stats = {}
if 'dosage_numeric' in df_processed.columns:
    dosage_stats['overall_mean'] = df_processed['dosage_numeric'].mean()
    dosage_stats['overall_std'] = df_processed['dosage_numeric'].std()
    dosage_stats['overall_p01'] = df_processed['dosage_numeric'].quantile(0.01)
    dosage_stats['overall_p99'] = df_processed['dosage_numeric'].quantile(0.99)
    # Calculate z-score and out-of-range flag *after* calculating stats
    mean = dosage_stats['overall_mean']
    std = dosage_stats['overall_std']
    p01 = dosage_stats['overall_p01']
    p99 = dosage_stats['overall_p99']
    df_processed['dosage_zscore_overall'] = (df_processed['dosage_numeric'] - mean) / std if std != 0 else 0
    df_processed['isDosageOutOfRange_overall'] = ~df_processed['dosage_numeric'].between(p01, p99, inclusive='both')
    print(f"  Calculated Dosage Normality. Stats: Mean={mean:.2f}, Std={std:.2f}, P01={p01:.2f}, P99={p99:.2f}")
else:
    print("    'dosage_numeric' column not found, setting default dosage stats.")
    dosage_stats = {'overall_mean': 0, 'overall_std': 1, 'overall_p01': 0, 'overall_p99': 1}
    df_processed['dosage_zscore_overall'] = 0
    df_processed['isDosageOutOfRange_overall'] = False

# Impute the newly created boolean dosage flag
if 'isDosageOutOfRange_overall' in df_processed.columns:
     df_processed['isDosageOutOfRange_overall'] = df_processed['isDosageOutOfRange_overall'].fillna(False)
     print("  Filled NaNs in 'isDosageOutOfRange_overall' with False.")


# --- 8. Split Data ---
print("\nSplitting data into train and validation sets...")
train_df, val_df = train_test_split(df_processed, test_size=VALIDATION_SPLIT_SIZE, random_state=42)
print(f"  Train set shape: {train_df.shape}")
print(f"  Validation set shape: {val_df.shape}")
# del df_processed # Removed comment

# --- 9. Preprocessing Pipeline ---
print("\nDefining and fitting preprocessing pipeline...")

# Load full list of classes for MLBs
print(f"  Loading frontend options from {OPTIONS_JSON_FILE} for MLB classes...")
try:
    with open(OPTIONS_JSON_FILE, 'r') as f:
        frontend_options = json.load(f)
    all_known_allergies_classes = frontend_options.get('knownAllergies', [])
    all_conditions_during_classes = frontend_options.get('conditions_during_encounter', [])
    all_new_conditions_classes = frontend_options.get('new_conditions_after_med_start', [])
    all_trial_sideeffects_classes = frontend_options.get('trialSideEffects', [])

    def get_classes_from_data(df, col_name):
        if col_name not in df.columns: return []
        all_items = set()
        df[col_name].apply(lambda lst: all_items.update(lst))
        return sorted(list(all_items))

    if not all_known_allergies_classes: all_known_allergies_classes = get_classes_from_data(train_df, 'knownAllergies')
    if not all_conditions_during_classes: all_conditions_during_classes = get_classes_from_data(train_df, 'conditions_during_encounter')
    if not all_new_conditions_classes: all_new_conditions_classes = get_classes_from_data(train_df, 'new_conditions_after_med_start')
    if not all_trial_sideeffects_classes: all_trial_sideeffects_classes = get_classes_from_data(train_df, 'trialSideEffects')

    if not all_known_allergies_classes: print("    Warning: No classes found or derived for knownAllergies.")
    if not all_conditions_during_classes: print("    Warning: No classes found or derived for conditions_during_encounter.")
    if not all_new_conditions_classes: print("    Warning: No classes found or derived for new_conditions_after_med_start.")
    if not all_trial_sideeffects_classes: print("    Warning: No classes found or derived for trialSideEffects.")

    print("    MLB class loading/derivation complete.")
except Exception as e:
    print(f"    Error loading/deriving options for MLB: {e}. MLBs might be incomplete.")
    all_known_allergies_classes, all_conditions_during_classes, all_new_conditions_classes, all_trial_sideeffects_classes = [], [], [], []


# Define feature groups (Updated based on feedback)
numerical_features = [
    'age', 'trialDuration', 'dosage_numeric', 'drugName_freq',
    'noteLength', 'noteWordCount', 'proc_count', 'claim_count',
    'dosage_zscore_overall',
    'avg_temp_during_trial', 'max_hr_during_trial',
    'min_bp_systolic_during_trial', 'max_bp_systolic_during_trial',
    'min_bp_diastolic_during_trial', 'max_bp_diastolic_during_trial',
    'count_side_effect_keywords_obs', 'count_improvement_keywords_obs', 'count_worsening_keywords_obs',
    # Add new numerical features
    'avg_condition_duration', 'std_temp_during_trial', 'std_hr_during_trial',
    'std_bp_systolic_during_trial', 'std_bp_diastolic_during_trial',
    # Add NLP features
    'nlp_verb_count', 'nlp_noun_count', 'nlp_adj_count', 'nlp_ner_count',
    # Add sentiment score
    'sentiment_score',
    # Add interaction feature
    'dosage_per_age'
]
categorical_features = [
    'gender',
    'sideEffectSeverity', 'overallHealthStatus', 'symptomImprovementScore',
    'isDosageOutOfRange_overall',
    'has_impossible_observation'
]
text_feature = 'doctorNotes'
multilabel_features = [
    'knownAllergies', 'conditions_during_encounter',
    'new_conditions_after_med_start', 'trialSideEffects'
]
multilabel_classes = {
    'knownAllergies': all_known_allergies_classes,
    'conditions_during_encounter': all_conditions_during_classes,
    'new_conditions_after_med_start': all_new_conditions_classes,
    'trialSideEffects': all_trial_sideeffects_classes
}

# Ensure all defined features actually exist in the training dataframe before passing to transformers
print("  Verifying feature columns exist in training data...")
train_cols = train_df.columns.tolist()
numerical_features = [f for f in numerical_features if f in train_cols]
categorical_features = [f for f in categorical_features if f in train_cols]
multilabel_features = [f for f in multilabel_features if f in train_cols]
if text_feature and text_feature not in train_cols:
    print(f"  Warning: Text feature '{text_feature}' not found in training data.")
    text_feature = None

print(f"    Using numerical features: {numerical_features}")
print(f"    Using categorical features: {categorical_features}")
print(f"    Using text feature: {text_feature}")
print(f"    Using multilabel features: {multilabel_features}")

# Create transformers
numeric_transformer = Pipeline([('imputer', SimpleImputer(strategy='median')), ('scaler', StandardScaler())])
categorical_transformer = Pipeline([('imputer', SimpleImputer(strategy='most_frequent')), ('onehot', OneHotEncoder(handle_unknown='ignore', sparse_output=False))])
text_transformer = TfidfVectorizer(max_features=TFIDF_MAX_FEATURES, stop_words='english')

# Define transformers list dynamically based on available features
transformers_list = []
if numerical_features: transformers_list.append(('num', numeric_transformer, numerical_features))
if categorical_features: transformers_list.append(('cat', categorical_transformer, categorical_features))
if text_feature: transformers_list.append(('txt', text_transformer, text_feature))

# Instantiate ColumnTransformer
if not transformers_list:
     print("Error: No features available for ColumnTransformer. Exiting.")
     exit()
ct = ColumnTransformer(transformers=transformers_list, remainder='drop')

# Fit ColumnTransformer on training data
print("  Fitting ColumnTransformer on training data...")
X_train_ct = ct.fit_transform(train_df)
print(f"    ColumnTransformer fit complete. Shape after CT: {X_train_ct.shape}")

# Fit MLBs separately using the potentially derived class lists
print("  Fitting MultiLabelBinarizers...")
mlb_transformers = {}
X_train_mlb_features = []
for col in multilabel_features:
    classes = multilabel_classes.get(col, [])
    if not classes:
        print(f"    Skipping MLB for '{col}' - no classes defined or derived.")
        continue
    if col not in train_df.columns:
        print(f"    Skipping MLB for '{col}' - column not found in training data.")
        continue
    mlb = MultiLabelBinarizer(classes=classes)
    try:
        mlb.fit([classes])
        transformed_data = mlb.transform(train_df[col].apply(lambda x: [item for item in x if item in mlb.classes_]))
        X_train_mlb_features.append(transformed_data)
        mlb_transformers[col] = mlb
        print(f"    Fitted and transformed MLB for '{col}'. Features: {len(mlb.classes_)}")
    except Exception as e:
        print(f"    Error with MLB for '{col}': {e}")
        X_train_mlb_features.append(np.zeros((len(train_df), len(classes))))

# Combine features
print("  Combining processed features...")
if hasattr(X_train_ct, "toarray"): X_train_ct = X_train_ct.toarray()
X_train_mlb_features_dense = [arr.toarray() if hasattr(arr, "toarray") else arr for arr in X_train_mlb_features]
X_train_processed = np.hstack([X_train_ct] + X_train_mlb_features_dense)
print(f"  Combined training features. Final shape: {X_train_processed.shape}")

# Save the fitted preprocessor components (NOW including dosage/freq stats)
preprocessor_artifact = {
    'col_transformer': ct,
    'mlb_transformers': mlb_transformers,
    'mlb_classes': multilabel_classes,
    'dosage_stats': dosage_stats,
    'drug_freq_map': drug_freq_map
}
print("  Preprocessor fitting complete.")

# --- 10. Train Models ---
print("\nTraining anomaly detection models...")

# --- Define Autoencoder ---
def build_autoencoder(input_dim, encoding_dim=32):
    """Builds a simple dense autoencoder."""
    # Encoder
    input_layer = keras.Input(shape=(input_dim,))
    encoded = layers.Dense(128, activation='relu')(input_layer)
    encoded = layers.Dense(64, activation='relu')(encoded)
    encoded = layers.Dense(encoding_dim, activation='relu')(encoded) # Bottleneck

    # Decoder
    decoded = layers.Dense(64, activation='relu')(encoded)
    decoded = layers.Dense(128, activation='relu')(decoded)
    decoded = layers.Dense(input_dim, activation='sigmoid')(decoded) # Output layer matches input

    autoencoder = keras.Model(input_layer, decoded)
    return autoencoder

# --- Train Sklearn Models ---
models = {
    'IsolationForest': IsolationForest(n_estimators=100, contamination='auto', random_state=42, n_jobs=-1), # Keep default for now
    'LocalOutlierFactor': LocalOutlierFactor(n_neighbors=20, contamination=0.15, novelty=True, n_jobs=-1), # Best contamination from tuning
    'OneClassSVM': OneClassSVM(gamma=1.0, nu=0.05) # New best gamma and nu from expanded tuning
}
trained_models = {}
for name, model in models.items():
    print(f"  Training {name}...")
    try:
        model.fit(X_train_processed)
        trained_models[name] = model
        print(f"    {name} training complete.")
    except Exception as e: print(f"    Error training {name}: {e}")

# --- Train Autoencoder ---
print("  Training Autoencoder...")
input_dim = X_train_processed.shape[1]
encoding_dim = max(16, input_dim // 8) # Example encoding dimension
autoencoder = build_autoencoder(input_dim, encoding_dim)
autoencoder.compile(optimizer='adam', loss='mse')

# Train on the processed training data
# Note: Consider normalizing data to [0, 1] if using sigmoid output, but StandardScaler output is fine for relu/mse
history = autoencoder.fit(X_train_processed, X_train_processed,
                          epochs=50, # Adjust epochs as needed
                          batch_size=256,
                          shuffle=True,
                          validation_split=0.1, # Use a portion of training data for validation during training
                          verbose=0) # Set verbose=1 or 2 to see training progress

print("    Autoencoder training complete.")

# --- Calculate Reconstruction Error Threshold ---
print("  Calculating reconstruction error threshold...")
train_reconstructions = autoencoder.predict(X_train_processed, verbose=0)
train_mse = np.mean(np.power(X_train_processed - train_reconstructions, 2), axis=1)
# Use mean + 3 standard deviations as threshold (adjust as needed)
threshold = np.mean(train_mse) + 3 * np.std(train_mse)
print(f"    Reconstruction error threshold set to: {threshold:.6f}")

# --- Save Autoencoder Model Separately ---
AUTOENCODER_MODEL_FILE = 'autoencoder_model.keras'
print(f"  Saving Autoencoder model to {AUTOENCODER_MODEL_FILE}...")
try:
    autoencoder.save(AUTOENCODER_MODEL_FILE)
    print("    Autoencoder model saved successfully.")
except Exception as e:
    print(f"    Error saving Autoencoder model: {e}")
    AUTOENCODER_MODEL_FILE = None # Indicate failure

# --- 11. Save Artifacts (including AE path and threshold) ---
print(f"\nSaving preprocessor, models, AE path, and threshold to {OUTPUT_ARTIFACTS_FILE}...")
artifacts_to_save = {
    'preprocessor': preprocessor_artifact,
    'models': trained_models,
    'autoencoder_model_path': AUTOENCODER_MODEL_FILE, # Store path again
    'autoencoder_threshold': threshold
}
try:
    joblib.dump(artifacts_to_save, OUTPUT_ARTIFACTS_FILE)
    print("  Artifacts saved successfully.")
except Exception as e: print(f"  Error saving artifacts: {e}")

# --- Validation code removed ---
# (Section 12: Analyze Thresholds on Validation Set was removed)

print("\nModel training script finished.")
```

```python
import pandas as pd
import numpy as np
import joblib
import ast
# import re # Removed unused import
import warnings
import json
import os # Added for environment variables
from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer # Import VADER
import tensorflow as tf
from tensorflow import keras
from keras.models import model_from_json # Import for loading AE config
# from sklearn.pipeline import Pipeline # Removed unused import

# Suppress warnings and TF logs
warnings.filterwarnings('ignore')
os.environ['TF_CPP_MIN_LOG_LEVEL'] = '2'
tf.get_logger().setLevel('ERROR')

# --- Configuration ---
# Use environment variable for artifacts path, defaulting to the tuned version
ARTIFACTS_FILE = os.getenv('ARTIFACTS_FILE', 'model_artifacts_tuned.joblib')

# --- Helper Functions ---
def parse_list_string(s):
    """Safely parses a string representation of a list or returns list if input is list."""
    try:
        if pd.isna(s): return []
        if isinstance(s, list): return s # Already a list
        if isinstance(s, str):
            # Handle simple comma-separated string if not a valid list literal
            if not (s.startswith('[') and s.endswith(']')):
                 s_cleaned = s.strip().strip("'\"")
                 # Return list of strings, handle empty strings after split
                 return [item.strip() for item in s_cleaned.split(',') if item.strip()]
            # If it looks like a list literal, try evaluating safely
            parsed_list = ast.literal_eval(s)
            if isinstance(parsed_list, list): return parsed_list
            elif isinstance(parsed_list, tuple): return list(parsed_list)
            else: return [str(parsed_list)] # Wrap single items
        return [] # Fallback for other types
    except (ValueError, SyntaxError, TypeError):
        # Fallback for strings that look like lists but aren't valid literals
        if isinstance(s, str) and s.strip():
             s_cleaned = s.strip().strip('[]\'"')
             return [item.strip() for item in s_cleaned.split(',') if item.strip()]
        return []

# --- Sentiment Analysis Function (copied from training script) ---
SIA = SentimentIntensityAnalyzer()
def get_sentiment_score(text):
    """Calculates the VADER compound sentiment score."""
    if pd.isna(text):
        return 0.0 # Neutral score for missing text
    try:
        # Ensure text is string
        text = str(text)
        vs = SIA.polarity_scores(text)
        return vs['compound']
    except Exception:
         # Handle potential errors during analysis
         return 0.0

# --- 1. Load Artifacts ---
print(f"Loading model artifacts from {ARTIFACTS_FILE}...")
try:
    artifacts = joblib.load(ARTIFACTS_FILE)
    preprocessor_artifact = artifacts['preprocessor']
    ct = preprocessor_artifact['col_transformer']
    mlb_transformers = preprocessor_artifact['mlb_transformers']
    mlb_classes = preprocessor_artifact.get('mlb_classes', {})
    dosage_stats = preprocessor_artifact.get('dosage_stats', {}) # Load dosage stats
    drug_freq_map = preprocessor_artifact.get('drug_freq_map', {}) # Load drug freq map
    models = artifacts['models']
    # Load Autoencoder path and threshold
    autoencoder_model_path = artifacts.get('autoencoder_model_path')
    autoencoder_threshold = artifacts.get('autoencoder_threshold')
    autoencoder = None
    if autoencoder_model_path and autoencoder_threshold is not None and os.path.exists(autoencoder_model_path):
        try:
            autoencoder = keras.models.load_model(autoencoder_model_path)
            print(f"  Autoencoder model loaded successfully from {autoencoder_model_path}.")
        except Exception as ae_load_err:
            print(f"  Warning: Error loading Autoencoder model from {autoencoder_model_path}: {ae_load_err}")
            autoencoder = None # Ensure it's None if loading fails
    else:
         print(f"  Warning: Autoencoder model path ('{autoencoder_model_path}') or threshold not found/invalid in artifacts, or file does not exist.")
         autoencoder = None # Ensure AE is None if path/threshold missing or file not found

    print("  Artifacts (preprocessor, models, stats, maps, AE path, AE threshold) loaded successfully.")
except FileNotFoundError:
    print(f"Error: Artifacts file not found - {ARTIFACTS_FILE}. Run train_anomaly_model.py first.")
    exit()
except KeyError as e:
    print(f"Error: Missing key in artifact file '{ARTIFACTS_FILE}': {e}. Ensure training script saves all components.")
    exit()
except Exception as e:
    print(f"Error loading artifacts: {e}. Cannot make predictions.")
    exit()

# --- Define Feature Lists (Matching Training Script V3 + New Features) ---
# These should align with the features *after* engineering in the training script
numerical_features = [
    'age', 'trialDuration', 'dosage_numeric', 'drugName_freq',
    'noteLength', 'noteWordCount', 'proc_count', 'claim_count',
    'dosage_zscore_overall',
    'avg_temp_during_trial', 'max_hr_during_trial',
    'min_bp_systolic_during_trial', 'max_bp_systolic_during_trial',
    'min_bp_diastolic_during_trial', 'max_bp_diastolic_during_trial',
    'count_side_effect_keywords_obs', 'count_improvement_keywords_obs', 'count_worsening_keywords_obs',
    # Added features from feature engineering step
    'avg_condition_duration', 'std_temp_during_trial', 'std_hr_during_trial',
    'std_bp_systolic_during_trial', 'std_bp_diastolic_during_trial',
    # Added NLP features
    'nlp_verb_count', 'nlp_noun_count', 'nlp_adj_count', 'nlp_ner_count',
    # Add sentiment score
    'sentiment_score',
    # Add interaction feature
    'dosage_per_age'
]
categorical_features = [
    'gender',
    'sideEffectSeverity', 'overallHealthStatus', 'symptomImprovementScore',
    'isDosageOutOfRange_overall',
    'has_impossible_observation'
]
text_feature = 'doctorNotes'
multilabel_features = [
    'knownAllergies', 'conditions_during_encounter',
    'new_conditions_after_med_start', 'trialSideEffects'
]

# Get the exact feature list expected by the loaded ColumnTransformer for non-MLB features
try:
    ct_expected_features = []
    for name, transformer_obj, columns in ct.transformers_:
        if transformer_obj != 'drop' and transformer_obj != 'passthrough':
             ct_expected_features.extend(columns)
    print(f"  Loaded ColumnTransformer expects features: {ct_expected_features}")
except Exception as e:
    print(f"  Warning: Could not reliably get features from loaded ColumnTransformer: {e}. Prediction might fail.")
    ct_expected_features = [] # Cannot proceed reliably without knowing expected features

# --- 2. Define Rule-Based Checks ---
def check_rules(record):
    """Applies predefined rules to the input record (using V3 feature names)."""
    rule_score = 0
    triggered_rules = []

    # Rule 1: Age out of range
    try: age = float(record.get('age', np.nan))
    except (ValueError, TypeError): age = np.nan
    if pd.notna(age) and not (0 <= age <= 120):
        rule_score += 1; triggered_rules.append("Age out of range (0-120)")

    # Rule 2: Trial duration invalid
    try: duration = float(record.get('trialDuration', np.nan))
    except (ValueError, TypeError): duration = np.nan
    if pd.notna(duration) and duration <= 0:
        rule_score += 1; triggered_rules.append("Trial duration <= 0")

    # Rule 3: Severity vs Side Effects mismatch (using trialSideEffects)
    severity = record.get('sideEffectSeverity', 'None')
    effects = record.get('trialSideEffects', [])
    if not isinstance(effects, list): effects = parse_list_string(effects) # Ensure list
    if severity in ['Severe', 'Moderate'] and (not effects or effects == ['None'] or not any(e != "Observation Keyword Trigger" for e in effects)): # Check if only trigger or empty
        rule_score += 1; triggered_rules.append("Severity is Moderate/Severe but no specific trial side effects listed")

    # Rule 4: Illogical Score vs Status
    try: score = float(record.get('symptomImprovementScore', 5))
    except (ValueError, TypeError): score = 5
    status = record.get('overallHealthStatus', 'Same')
    if score >= 8 and status == 'Worse':
        rule_score += 1; triggered_rules.append("High symptom improvement score but overall status is Worse")
    if score <= 2 and status == 'Improved':
        rule_score += 1; triggered_rules.append("Low symptom improvement score but overall status is Improved")

    # Rule 5: Gender vs Condition (using conditions_during_encounter)
    gender = record.get('gender', 'Unknown')
    conditions = record.get('conditions_during_encounter', [])
    if not isinstance(conditions, list): conditions = parse_list_string(conditions)
    pregnancy_conditions = {'Normal Pregnancy', 'Antepartum Eclampsia', 'Preeclampsia', 'Tubal Pregnancy'}
    if gender == 'Male' and any(cond in pregnancy_conditions for cond in conditions):
         rule_score += 1; triggered_rules.append("Pregnancy-related condition listed for Male patient")

    # Rule 6: Unusually Long Trial Duration
    if pd.notna(duration) and duration > (365 * 5):
        rule_score += 1; triggered_rules.append(f"Trial duration seems excessively long (> {365*5} days)")

    # Rule 8: Impossible Observation Flag
    if record.get('has_impossible_observation', False):
        rule_score += 1; triggered_rules.append("Record contains physiologically impossible observation value")

    # Rule 9: Dosage Out of Range Flag (calculated during prediction)
    if record.get('isDosageOutOfRange_overall', False):
        rule_score += 1; triggered_rules.append("Dosage is outside typical range (1st-99th percentile)")

    return rule_score, triggered_rules

# --- 3. Define Prediction Function ---
def predict_report_anomaly(record_dict):
    """
    Predicts the anomaly score for a single report record using V3 artifacts.
    Assumes record_dict contains features engineered similarly to the training script.
    Calculates dosage normality and drug frequency using loaded artifacts.
    """
    print(f"\nProcessing record...")

    # --- Prepare DataFrame from Input ---
    # Create a DataFrame directly from the input dictionary
    input_df = pd.DataFrame([record_dict]) # Create DataFrame with one row

    # --- Feature Engineering (Prediction Time) ---
    # Ensure correct types before calculations
    # Use .get() on the original dict for safety before assigning to DataFrame
    input_df['dosage_numeric'] = pd.to_numeric(record_dict.get('dosage_numeric'), errors='coerce')
    input_df['drugName'] = str(record_dict.get('drugName', 'Unknown')) # Ensure drugName exists as string
    input_df['doctorNotes'] = str(record_dict.get('doctorNotes', '')) # Ensure doctorNotes exists as string

    # 1. Sentiment Score
    input_df['sentiment_score'] = get_sentiment_score(input_df['doctorNotes'].iloc[0])
    print(f"  Calculated sentiment score: {input_df['sentiment_score'].iloc[0]:.4f}")

    # 2. Dosage per Age
    input_df['age'] = pd.to_numeric(record_dict.get('age'), errors='coerce') # Ensure age is numeric
    if 'dosage_numeric' in input_df.columns and 'age' in input_df.columns:
        # Replace age 0 or NaN with NaN to avoid division error, then fillna
        input_df['dosage_per_age'] = (input_df['dosage_numeric'] / input_df['age'].replace(0, np.nan)).fillna(0)
        print(f"  Calculated dosage_per_age: {input_df['dosage_per_age'].iloc[0]:.4f}")
    else:
        input_df['dosage_per_age'] = 0
        print("  Warning: Could not calculate dosage_per_age (missing dosage or age).")


    # 3. Dosage Normality (using loaded stats)
    if dosage_stats and 'dosage_numeric' in input_df.columns:
        mean = dosage_stats.get('overall_mean', 0)
        std = dosage_stats.get('overall_std', 1)
        p01 = dosage_stats.get('overall_p01', 0)
        p99 = dosage_stats.get('overall_p99', 1)
        if std == 0: std = 1

        input_df['dosage_numeric'] = input_df['dosage_numeric'].fillna(mean) # Impute NaN dosage
        input_df['dosage_zscore_overall'] = (input_df['dosage_numeric'] - mean) / std
        input_df['isDosageOutOfRange_overall'] = ~input_df['dosage_numeric'].between(p01, p99, inclusive='both')
        print("  Calculated dosage normality features using loaded stats.")
    else:
        input_df['dosage_zscore_overall'] = 0
        input_df['isDosageOutOfRange_overall'] = False
        print("  Warning: Dosage stats/dosage_numeric missing. Using default normality features.")

    # 3. Drug Frequency (using loaded map)
    if drug_freq_map and 'drugName' in input_df.columns:
        # Use .get on the map for unknown drugs
        input_df['drugName_freq'] = input_df['drugName'].apply(lambda x: drug_freq_map.get(x, 0.0))
        print("  Calculated drug frequency using loaded map.")
    else:
        input_df['drugName_freq'] = 0.0
        print("  Warning: Drug frequency map/drugName missing. Using 0.0.")

    # Ensure list columns are lists and exist
    for col in multilabel_features:
        input_value = record_dict.get(col, []) # Get value from original dict
        input_df[col] = [parse_list_string(input_value)] # Apply parsing and assign as list within list

    # Add other potentially missing columns needed by the pipeline with defaults
    all_pipeline_features = ct_expected_features + multilabel_features
    missing_cols = []
    for feature in all_pipeline_features:
         if feature not in input_df.columns:
             missing_cols.append(feature)
             default_value = None
             if feature in numerical_features: default_value = 0
             elif feature in categorical_features: default_value = 'Unknown'
             elif feature == text_feature: default_value = ''
             # Defaults for specific known features if needed (add new ones)
             if feature == 'age': default_value = 30
             if feature == 'trialDuration': default_value = 30
             if feature == 'avg_condition_duration': default_value = 0 # Example default
             if feature == 'std_temp_during_trial': default_value = 0
             if feature == 'std_hr_during_trial': default_value = 0
             if feature == 'std_bp_systolic_during_trial': default_value = 0
             if feature == 'std_bp_diastolic_during_trial': default_value = 0
             # Add defaults for NLP features
             if feature == 'nlp_verb_count': default_value = 0
             if feature == 'nlp_noun_count': default_value = 0
             if feature == 'nlp_adj_count': default_value = 0
             if feature == 'nlp_ner_count': default_value = 0
             if feature == 'symptomImprovementScore': default_value = 5
             if feature == 'sideEffectSeverity': default_value = 'None'
             if feature == 'overallHealthStatus': default_value = 'Same'
             if feature == 'gender': default_value = 'Other'
             if feature == 'has_impossible_observation': default_value = False
             input_df[feature] = default_value
    if missing_cols:
             if feature == 'dosage_per_age': default_value = 0 # Default for interaction feature
             input_df[feature] = default_value
    if missing_cols:
        print(f"    Warning: Added default values for missing features: {missing_cols}")

    # Apply Rule Checks (after calculating derived features)
    # Convert Series back to dict for check_rules
    record_for_rules = input_df.iloc[0].to_dict()
    # Ensure list columns are simple lists for rule checking
    for col in multilabel_features:
        if col in record_for_rules and isinstance(record_for_rules[col], list) and len(record_for_rules[col]) == 1 and isinstance(record_for_rules[col][0], list):
             record_for_rules[col] = record_for_rules[col][0] # Extract inner list
        elif col in record_for_rules and not isinstance(record_for_rules[col], list):
             record_for_rules[col] = parse_list_string(record_for_rules[col]) # Ensure it's a list

    rule_score, triggered_rules = check_rules(record_for_rules)
    print(f"  Rule Check Score: {rule_score}, Triggered: {triggered_rules}")

    # --- Preprocessing using loaded artifacts ---
    # Separate MLB processing
    mlb_processed_arrays = []
    for col in multilabel_features:
        if col in mlb_transformers:
            mlb = mlb_transformers[col]
            mlb_classes_for_col = mlb_classes.get(col, [])
            if not mlb_classes_for_col: continue
            try:
                data_to_transform = input_df[col].tolist()
                mlb.classes_ = mlb_classes_for_col
                # Filter input data to only include known classes before transform
                data_filtered = [[item for item in sublist if item in mlb.classes_] for sublist in data_to_transform]
                transformed = mlb.transform(data_filtered)
                mlb_processed_arrays.append(transformed)
                print(f"  Applied MLB for {col}")
            except Exception as e:
                 print(f"  Error applying MLB for {col}: {e}. Adding zeros.")
                 mlb_processed_arrays.append(np.zeros((len(input_df), len(mlb_classes_for_col))))
        else:
             mlb_classes_for_col = mlb_classes.get(col, [])
             mlb_processed_arrays.append(np.zeros((len(input_df), len(mlb_classes_for_col))))

    # Prepare DataFrame for ColumnTransformer
    input_df_for_ct = input_df.copy()
    missing_ct_cols = []
    if not ct_expected_features:
         print("Error: Cannot determine expected features for ColumnTransformer.")
         return {"error": "Preprocessing failed (CT feature definition error)", "rule_score": rule_score, "triggered_rules": triggered_rules}

    for feature in ct_expected_features:
        if feature not in input_df_for_ct.columns:
             missing_ct_cols.append(feature)
             if feature in numerical_features: input_df_for_ct[feature] = 0
             elif feature in categorical_features: input_df_for_ct[feature] = 'Unknown'
             elif feature == text_feature: input_df_for_ct[feature] = ''
             else: input_df_for_ct[feature] = None
    if missing_ct_cols:
        print(f"    Warning: Added placeholder columns for missing CT features: {missing_ct_cols}")

    # Select and reorder columns based on the predefined lists used for training
    # This ensures the order matches how the ColumnTransformer was fitted
    final_ct_cols = []
    if numerical_features: final_ct_cols.extend(numerical_features)
    if categorical_features: final_ct_cols.extend(categorical_features)
    if text_feature: final_ct_cols.append(text_feature)

    # Ensure all columns in final_ct_cols exist in input_df_for_ct, adding defaults if needed
    missing_final_cols = []
    for col in final_ct_cols:
        if col not in input_df_for_ct.columns:
            missing_final_cols.append(col)
            # Add appropriate placeholder based on expected type
            if col in numerical_features: input_df_for_ct[col] = 0
            elif col in categorical_features: input_df_for_ct[col] = 'Unknown'
            elif col == text_feature: input_df_for_ct[col] = ''
            else: input_df_for_ct[col] = None
    if missing_final_cols:
         print(f"    Warning: Added default values for missing final CT columns: {missing_final_cols}")

    try:
        input_df_for_ct = input_df_for_ct[final_ct_cols] # Select columns in the correct order
        print(f"  Prepared DataFrame for ColumnTransformer with columns: {list(input_df_for_ct.columns)}")
    except KeyError as e:
        print(f"  Error selecting final columns for ColumnTransformer: {e}")
        return {"error": f"Missing expected column during final selection: {e}", "rule_score": rule_score, "triggered_rules": triggered_rules}

    # Apply ColumnTransformer
    try:
        # Explicitly clean the text feature column right before transformation
        if text_feature and text_feature in input_df_for_ct.columns:
            cleaned_notes = []
            for note in input_df_for_ct[text_feature]:
                if isinstance(note, str):
                    cleaned_notes.append(note)
                else: # Handles None, NaN, and any other non-string types
                    cleaned_notes.append('')
            input_df_for_ct[text_feature] = cleaned_notes
            print(f"  Explicitly cleaned '{text_feature}' before CT.")

        X_input_ct = ct.transform(input_df_for_ct)
        if hasattr(X_input_ct, "toarray"): X_input_ct = X_input_ct.toarray()
        print(f"  ColumnTransformer successful. Shape: {X_input_ct.shape}")
    except Exception as e:
        print(f"  Error during ColumnTransformer transformation: {e}")
        return {"error": "Preprocessing failed (CT)", "rule_score": rule_score, "triggered_rules": triggered_rules}

    # Combine CT features and MLB features
    try:
        if X_input_ct.shape[0] == 0 and any(arr.shape[0] > 0 for arr in mlb_processed_arrays):
             X_input_ct = np.zeros((mlb_processed_arrays[0].shape[0], X_input_ct.shape[1]))
        elif any(arr.shape[0] != X_input_ct.shape[0] for arr in mlb_processed_arrays):
             raise ValueError("Mismatch in number of rows between CT and MLB outputs.")

        X_input_processed = np.hstack([X_input_ct] + mlb_processed_arrays)
        print(f"  Combined all features. Final shape: {X_input_processed.shape}")
    except ValueError as e:
         print(f"  Error combining features: {e}")
         print(f"    CT shape: {X_input_ct.shape}")
         for i, arr in enumerate(mlb_processed_arrays): print(f"    MLB {i} shape: {arr.shape}")
         return {"error": "Feature combination failed", "rule_score": rule_score, "triggered_rules": triggered_rules}

    # --- Get predictions from all models ---
    predictions = {}
    for name, model in models.items():
        try:
            score = np.nan
            label = "Error"
            pred = model.predict(X_input_processed)[0]
            label = "Outlier/Anomaly" if pred == -1 else "Inlier/Normal"

            if hasattr(model, "decision_function"):
                score = model.decision_function(X_input_processed)[0]
            elif hasattr(model, "score_samples"):
                score = model.score_samples(X_input_processed)[0]

            predictions[name] = {"raw_score": float(score), "prediction_label": label}
            print(f"  {name} Score: {score:.4f}, Prediction: {label}")
        except Exception as e:
            print(f"  Error getting prediction for {name}: {e}")
            predictions[name] = {"error": str(e)}

    # --- Get Autoencoder Prediction ---
    if autoencoder and autoencoder_threshold is not None:
        try:
            reconstructions = autoencoder.predict(X_input_processed, verbose=0)
            mse = np.mean(np.power(X_input_processed - reconstructions, 2), axis=1)
            ae_pred = -1 if mse[0] > autoencoder_threshold else 1
            ae_label = "Outlier/Anomaly" if ae_pred == -1 else "Inlier/Normal"
            predictions['Autoencoder'] = {"raw_score": float(mse[0]), "prediction_label": ae_label}
            print(f"  Autoencoder MSE: {mse[0]:.6f}, Threshold: {autoencoder_threshold:.6f}, Prediction: {ae_label}")
        except Exception as e:
            print(f"  Error getting prediction for Autoencoder: {e}")
            predictions['Autoencoder'] = {"error": str(e)}
    else:
        print("  Skipping Autoencoder prediction (model or threshold not loaded).")


    # --- Final Combined Decision ---
    final_prediction_label = "Error"
    final_prediction_numeric = 0 # Default to error/unknown

    # Prioritize rule violations
    if rule_score > 0:
        final_prediction_label = "Outlier/Anomaly"
        final_prediction_numeric = -1
    else:
        # Implement Majority Voting (including Autoencoder if available) if no rules triggered
        anomaly_votes = 0
        model_names = ['IsolationForest', 'LocalOutlierFactor', 'OneClassSVM']
        if 'Autoencoder' in predictions and 'error' not in predictions['Autoencoder']:
             model_names.append('Autoencoder') # Add AE to the vote if it worked

        valid_model_predictions = 0
        for name in model_names:
            if name in predictions and 'error' not in predictions[name]:
                valid_model_predictions += 1
                if predictions[name].get('prediction_label') == 'Outlier/Anomaly':
                    anomaly_votes += 1

        # Majority requires > half the votes (e.g., 2/3 or 3/4)
        if valid_model_predictions > 0 and anomaly_votes > (valid_model_predictions / 2.0):
            final_prediction_label = "Outlier/Anomaly"
            final_prediction_numeric = -1
            print(f"  Majority Vote Result: Anomaly ({anomaly_votes}/{valid_model_predictions} votes)")
        elif valid_model_predictions > 0: # If not majority anomaly, predict normal
            final_prediction_label = "Inlier/Normal"
            final_prediction_numeric = 1
            print(f"  Majority Vote Result: Normal ({anomaly_votes}/{valid_model_predictions} votes)")
        else:
            # Handle case where all models failed
            print("  Warning: Rules not triggered and no valid model predictions available. Defaulting final prediction to Normal.")
            final_prediction_label = "Inlier/Normal"
            final_prediction_numeric = 1


    # Return combined results
    return {
        "rule_score": rule_score,
        "triggered_rules": triggered_rules,
        "model_predictions": predictions,
        "final_combined_prediction": {
             "label": final_prediction_label,
             "numeric": final_prediction_numeric
        }
    }
```
